### PROBLEM FRAMING

The **main goal** of this project is to **develop a predictive model** that estimates that **a loan** application will be **approved** or **not** to support credit decision-making.

#### **SMART framing**

- Specific: Predict default risk at application time.
- Measurable: Achieve ROC-AUC ≥ 0.80 and reduce default rate by ≥ X% compared to rule-based baseline.
- Achievable: Use historical labeled loan data.
- Relevant: Default reduction directly improves portfolio profitability.
- Time-bound: Deploy model within 4–6 months.

#### **Benefits**

- Reduced default losses
- Faster approval process
- Scalable underwriting
- More consistent decisions
- Optimized risk-adjusted return

If average default loss per loan = L
If model reduces default rate by Δ%
Total annual savings ≈ Portfolio_Size × Δ% × L

Even a small % improvement produces large financial impact.

#### Costs

- Data engineering
- Model development
- Cloud infrastructure
- Monitoring & retraining
- Compliance and explainability requirements
- Drift detection maintenance
- Maintenance cost is ongoing but generally far smaller than avoided credit losses.
  
Of course, there are solutions **other than machine learning**. We have different alternatives such as : fixed income thresholds, rule-based scoring, logistic scorecard manually tuned, human underwriters

However, rules cannot capture nonlinear feature interactions, performance plateaus quickly and it's hard to adapt to distribution shifts.

So, solving loan approval problem is feasible without ML, but ML likely yields better discrimination power and financial optimization.

```python
# ============================================
# 1. IMPORTS
# ============================================

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.linear_model import (
    LogisticRegression,
    Ridge,
    Lasso,
    BayesianRidge,
    LinearRegression
)

from sklearn.discriminant_analysis import (
    LinearDiscriminantAnalysis,
    QuadraticDiscriminantAnalysis
)

from sklearn.naive_bayes import GaussianNB

from sklearn.metrics import roc_auc_score

# ============================================
# 2. LOAD DATA
# ============================================

train_df = pd.read_csv("./playground-series-s4e10/train.csv")
test_df = pd.read_csv("./playground-series-s4e10/test.csv")

print("Train shape:", train_df.shape)
print("Test shape:", test_df.shape)

# ============================================
# 3. DATA WRANGLING (ON TRAIN + TEST)
# ============================================

def preprocess_dataframe(df):
    
    df = df.copy()
    
    # Fill numeric missing
    df = df.fillna(df.median(numeric_only=True))
    
    # Fill categorical missing
    for col in df.select_dtypes(include="object"):
        df[col] = df[col].fillna(df[col].mode()[0])
    
    # Binary encoding
    if "cb_person_default_on_file" in df.columns:
        df["cb_person_default_on_file"] = df["cb_person_default_on_file"].map({"N":0, "Y":1})
    
    # Ordinal encoding
    grade_mapping = {
        "A":1, "B":2, "C":3, "D":4,
        "E":5, "F":6, "G":7
    }
    if "loan_grade" in df.columns:
        df["loan_grade"] = df["loan_grade"].map(grade_mapping)
    
    return df

train_df = preprocess_dataframe(train_df)
test_df = preprocess_dataframe(test_df)

# ============================================
# 4. SPLIT TRAIN INTO TRAIN + VALIDATION
# ============================================

X = train_df.drop(columns=["loan_status", "id"])
y = train_df["loan_status"]

X_train, X_val, y_train, y_val = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

# ============================================
# 5. PREPROCESSING PIPELINE
# ============================================

numeric_features = X.select_dtypes(include=np.number).columns
categorical_features = X.select_dtypes(include="object").columns

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features),
    ]
)

# ============================================
# 6. MODELS TO TEST
# ============================================

models = {
    "Logistic Regression": LogisticRegression(max_iter=2000),
    "LDA": LinearDiscriminantAnalysis(),
    "QDA": QuadraticDiscriminantAnalysis(),
    "GaussianNB": GaussianNB(),
    "Ridge": Ridge(),
    "Lasso": Lasso(),
    "Bayesian Ridge": BayesianRidge(),
    "Linear Regression": LinearRegression()
}

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

results = {}

print("\nCross-validation results:\n")

for name, model in models.items():
    
    pipe = Pipeline([
        ("preprocessing", preprocessor),
        ("model", model)
    ])
    
    try:
        scores = cross_val_score(pipe, X_train, y_train,
                                 cv=cv,
                                 scoring="roc_auc")
        
        results[name] = scores.mean()
        print(f"{name}: {scores.mean():.4f}")
    
    except:
        print(f"{name} failed.")

# ============================================
# 7. SELECT BEST MODEL
# ============================================

best_model_name = max(results, key=results.get)
print("\nBest model:", best_model_name)

best_model = models[best_model_name]

final_pipeline = Pipeline([
    ("preprocessing", preprocessor),
    ("model", best_model)
])

final_pipeline.fit(X_train, y_train)

# ============================================
# 8. VALIDATION PERFORMANCE
# ============================================

y_val_proba = final_pipeline.predict_proba(X_val)[:,1]
val_auc = roc_auc_score(y_val, y_val_proba)

print("\nValidation ROC-AUC:", val_auc)

# ============================================
# 9. TRAIN ON FULL TRAIN DATA
# ============================================

final_pipeline.fit(X, y)

# ============================================
# 10. FINAL PREDICTION ON TEST.CSV
# ============================================

X_test_final = test_df.drop(columns=["id"])

test_predictions = final_pipeline.predict_proba(X_test_final)[:,1]

submission = pd.DataFrame({
    "id": test_df["id"],
    "loan_status": test_predictions
})

submission.to_csv("submission.csv", index=False)

print("\nSubmission file created.")
```

```python
# ============================================
# 1. IMPORTS
# ============================================

import numpy as np
import pandas as pd
from scipy.special import expit  # sigmoid pour modèles de régression

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.linear_model import (
    LogisticRegression,
    Ridge,
    Lasso,
    BayesianRidge,
    LinearRegression
)

from sklearn.discriminant_analysis import (
    LinearDiscriminantAnalysis,
    QuadraticDiscriminantAnalysis
)

from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

# ============================================
# 2. LOAD DATA
# ============================================

train_df = pd.read_csv("./playground-series-s4e10/train.csv")
test_df = pd.read_csv("./playground-series-s4e10/test.csv")

print("Train shape:", train_df.shape)
print("Test shape:", test_df.shape)

# ============================================
# 3. DATA WRANGLING
# ============================================

def preprocess_dataframe(df):
    df = df.copy()
    
    # Fill numeric missing
    df = df.fillna(df.median(numeric_only=True))
    
    # Fill categorical missing
    for col in df.select_dtypes(include="object"):
        df[col] = df[col].fillna(df[col].mode()[0])
    
    # Binary encoding
    if "cb_person_default_on_file" in df.columns:
        df["cb_person_default_on_file"] = df["cb_person_default_on_file"].map({"N":0, "Y":1})
    
    # Ordinal encoding
    grade_mapping = {"A":1, "B":2, "C":3, "D":4, "E":5, "F":6, "G":7}
    if "loan_grade" in df.columns:
        df["loan_grade"] = df["loan_grade"].map(grade_mapping)
    
    return df

train_df = preprocess_dataframe(train_df)
test_df = preprocess_dataframe(test_df)

# ============================================
# 4. SPLIT TRAIN INTO TRAIN + VALIDATION
# ============================================

X = train_df.drop(columns=["loan_status", "id"])
y = train_df["loan_status"]

X_train, X_val, y_train, y_val = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

# ============================================
# 5. PREPROCESSING PIPELINE
# ============================================

numeric_features = X.select_dtypes(include=np.number).columns
categorical_features = X.select_dtypes(include="object").columns

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_features),
        ("cat", OneHotEncoder(drop='first', handle_unknown="ignore"), categorical_features),
    ]
)

# ============================================
# 6. MODELS SEPARÉS
# ============================================

def compute_metrics(y_true, y_proba, threshold=0.5):
    y_pred = (y_proba >= threshold).astype(int)
    return {
        "roc_auc": roc_auc_score(y_true, y_proba),
        "accuracy": accuracy_score(y_true, y_pred),
        "precision": precision_score(y_true, y_pred),
        "recall": recall_score(y_true, y_pred),
        "f1": f1_score(y_true, y_pred)
    }

# ============================================
# Logistic Regression
# ============================================
log_reg = Pipeline([
    ("preprocessing", preprocessor),
    ("model", LogisticRegression(max_iter=2000))
])
log_reg.fit(X_train, y_train)
y_val_proba_logreg = log_reg.predict_proba(X_val)[:,1]
metrics_logreg = compute_metrics(y_val, y_val_proba_logreg)
print("Logistic Regression metrics:", metrics_logreg)


# ============================================
# LDA
# ============================================
lda = Pipeline([
    ("preprocessing", preprocessor),
    ("model", LinearDiscriminantAnalysis())
])
lda.fit(X_train, y_train)
y_val_proba_lda = lda.predict_proba(X_val)[:,1]
metrics_lda = compute_metrics(y_val, y_val_proba_lda)
print("LDA metrics:", metrics_lda)


# ============================================
# QDA
# ============================================
qda = Pipeline([
    ("preprocessing", preprocessor),
    ("model", QuadraticDiscriminantAnalysis())
])
qda.fit(X_train, y_train)
y_val_proba_qda = qda.predict_proba(X_val)[:,1]
metrics_qda = compute_metrics(y_val, y_val_proba_qda)
print("QDA metrics:", metrics_qda)


# ============================================
# GaussianNB
# ============================================
gnb = Pipeline([
    ("preprocessing", preprocessor),
    ("model", GaussianNB())
])
gnb.fit(X_train, y_train)
y_val_proba_gnb = gnb.predict_proba(X_val)[:,1]
metrics_gnb = compute_metrics(y_val, y_val_proba_gnb)
print("GaussianNB metrics:", metrics_gnb)


# ============================================
# Ridge (régression -> probabilité)
# ============================================
ridge = Pipeline([
    ("preprocessing", preprocessor),
    ("model", Ridge())
])
ridge.fit(X_train, y_train)
y_val_proba_ridge = expit(ridge.predict(X_val))
metrics_ridge = compute_metrics(y_val, y_val_proba_ridge)
print("Ridge metrics:", metrics_ridge)


# ============================================
# Lasso (régression -> probabilité)
# ============================================
lasso = Pipeline([
    ("preprocessing", preprocessor),
    ("model", Lasso())
])
lasso.fit(X_train, y_train)
y_val_proba_lasso = expit(lasso.predict(X_val))
metrics_lasso = compute_metrics(y_val, y_val_proba_lasso)
print("Lasso metrics:", metrics_lasso)


# ============================================
# Bayesian Ridge (régression -> probabilité)
# ============================================
bayesian_ridge = Pipeline([
    ("preprocessing", preprocessor),
    ("model", BayesianRidge())
])
bayesian_ridge.fit(X_train, y_train)
y_val_proba_br = expit(bayesian_ridge.predict(X_val))
metrics_br = compute_metrics(y_val, y_val_proba_br)
print("Bayesian Ridge metrics:", metrics_br)


# ============================================
# Linear Regression (régression -> probabilité)
# ============================================
linear_reg = Pipeline([
    ("preprocessing", preprocessor),
    ("model", LinearRegression())
])
linear_reg.fit(X_train, y_train)
y_val_proba_lr = expit(linear_reg.predict(X_val))
metrics_lr = compute_metrics(y_val, y_val_proba_lr)
print("Linear Regression metrics:", metrics_lr)


# ============================================
# 7. SÉLECTION DU MEILLEUR MODÈLE
# ============================================

# Comparer tous les ROC-AUC
roc_scores = {
    "Logistic Regression": roc_auc_logreg,
    "LDA": roc_auc_lda,
    "QDA": roc_auc_qda,
    "GaussianNB": roc_auc_gnb,
    "Ridge": roc_auc_ridge,
    "Lasso": roc_auc_lasso,
    "Bayesian Ridge": roc_auc_br,
    "Linear Regression": roc_auc_lr
}

best_model_name = max(roc_scores, key=roc_scores.get)
print("\nBest model:", best_model_name)

# Re-créer le meilleur modèle pour l'entraînement final
best_model_pipeline = {
    "Logistic Regression": log_reg,
    "LDA": lda,
    "QDA": qda,
    "GaussianNB": gnb,
    "Ridge": ridge,
    "Lasso": lasso,
    "Bayesian Ridge": bayesian_ridge,
    "Linear Regression": linear_reg
}[best_model_name]

# ============================================
# 8. TRAIN ON FULL TRAIN DATA
# ============================================

best_model_pipeline.fit(X, y)

# ============================================
# 9. FINAL PREDICTION ON TEST.CSV
# ============================================

X_test_final = test_df.drop(columns=["id"])
y_test_proba = best_model_pipeline.predict_proba(X_test_final)[:,1] if hasattr(best_model_pipeline.named_steps['model'], "predict_proba") else expit(best_model_pipeline.predict(X_test_final))

submission = pd.DataFrame({
    "id": test_df["id"],
    "loan_status": y_test_proba
})

submission.to_csv("submission.csv", index=False)
print("\nSubmission file created.")
```
