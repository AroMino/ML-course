Minimize x_1+x_2+...+x_77

Constraints:

        x_1*nut_1_1+x_2*nut_2_1+...+x_77*nut_77_1 >= U_1
        
        x_1*nut_1_2+x_2*nut_2_2+...+x_77*nut_77_2 >= U_2
        
        ...
        
        x_1*nut_1_9+x_2*nut_2_9+...+x_77*nut_77_9 >= U_9

        

x vector (x_1, x_2, ... x_77) variable that represents dollar spent on each food

nut matrix that represents nutrients values for each food

U vector that represents minimum recommended intake

```python
import pandas as pd
import numpy as np
import cvxpy as cp
```

```python
# Load and process data
data = pd.read_csv("diet.csv")
arr = data.to_numpy() # transform in numpy array
nut = arr[:, 1:] # remove string cells
```

```python
# Minimum recommended intake
# | Nutrient | Minimum recommended intake |
# |---|---|
# | Calories | 3 kcal |
# | Protein | 70 g |
# | Calcium | 0.8 g |
# | Iron | 12 mg |
# | Vitamin A | 5000 IU |
# | Vitamin B1 | 1.8 mg |
# | Vitamin B2 | 2.7 mg |
# | Niacin | 18 mg |
# | Vitamin C | 75 mg |
U = np.array([3, 70, 0.8, 12, 5, 1.8, 2.7, 18, 75])
```

```python
# Problem
x = cp.Variable(arr.shape[0], nonneg=True)
objective = cp.Minimize(cp.sum(x))
constraints = [nut.T @ x >= U]
prob = cp.Problem(objective, constraints)
```

```python
# Solve
res = prob.solve()
print("Minimal prince :", round(res, 5))
```

```python
# Solutions
quantities = x.value

# Minimal spending per food
food_name = arr[:, 0]

for i, q in enumerate(quantities):
    print(f"{food_name[i]} : {round(q, 5)} $")
    
```
