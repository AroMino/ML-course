```python
import cvxpy as cp
import numpy as np

C = np.array([8, 10])

A = np.array([
    [0.4, 1.2],
    [6, 10],
    [0.4, 0.6]
])

b = np.array([70, 50, 12])

# Variable : quantité de pommes et oranges
x = cp.Variable(2, nonneg=True)

# Fonction objectif : minimiser le coût
objective = cp.Minimize(C @ x)

# Contraintes nutritionnelles
constraints = [
    A @ x >= b
]

# Problème
problem = cp.Problem(objective, constraints)
problem.solve()

print("Quantité pommes :", x.value[0])
print("Quantité oranges :", x.value[1])
print("Coût minimum :", problem.value)
```
