```python
import numpy as np
```

```python
def normalize(X):
    X = (X - np.mean(X, axis=0))
    return X
```

```python
class PCA:
    def __init__(self):
        pass

    def fit(self, X, l=None):
        self.l = l
        self.W = None
        
        N = X.shape[0]
        X_c = normalize(X)

        # Covariances
        covariances = (X_c.T @ X_c) / (N-1)

        # Valeurs propres et vecteurs propres
        eigenvalues, eigenvectors = np.linalg.eig(covariances)
        
        # Trier par valeurs propres décroissantes
        idx = np.argsort(eigenvalues)[::-1]
        eigenvalues = eigenvalues[idx]
        eigenvectors = eigenvectors[:, idx]

        # Garder seulement l si spécifié
        if self.l is not None:
            eigenvalues = eigenvalues[:self.l]
            eigenvectors = eigenvectors[:, :self.l]

        self.W = eigenvectors

    def transform(self, X):
        # Centrer les données
        X_c = normalize(X)

        # Projection sur les composantes principales
        X_pca = X_c @ self.W

        return X_pca
```

```python
pca = PCA()
X = np.random.rand(10, 4) * 10
pca.fit(X)
X_pca = pca.transform(X)
X
```
