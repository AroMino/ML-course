|---|---|

# 0. Problem Framing

> **Tâche :** Classification supervisée 3 classes (Low / Medium / High) sur 9618 entreprises africaines  
> **Métrique :** Weighted F1-score  
> **Déséquilibre :** Low 65% / Medium 30% / High 5% — la classe High est le principal enjeu  
> **Baseline à battre :** 0.8875 → Objectif : **0.92+**
>
> **Stratégie :**  
> 1. Target Encoding safe (CV) pour les catégorielles → réduit le bruit vs OHE  
> 2. Feature engineering financier avancé + interactions pays  
> 3. LightGBM sans class_weight + **sample_weight boost pour High**  
> 4. Stacking OOF (LightGBM + XGBoost + RandomForest → LogReg méta)  
> 5. Blending final avec poids optimisés

# 3. Preprocessing Pipeline

```python
# ── 3.1 Copie & nettoyage des valeurs catégorielles ────────────────────────
def clean_strings(df):
    df = df.copy()
    for col in df.select_dtypes('object').columns:
        if col in [ID_COL]: continue
        s = df[col].astype(str).str.lower().str.strip()
        # Standardiser les variantes de "don't know"
        s = s.str.replace(r"don[?'\u2019\u0301]t know(?: or n/a)?", "dont_know", regex=True)
        # Standardiser "have now" → 'have_now'
        s = s.str.replace('have now', 'have_now', regex=False)
        # Standardiser "never had" → 'never_had'
        s = s.str.replace('never had', 'never_had', regex=False)
        # Standardiser "used to have"
        s = s.str.replace(r'used to have.*', 'used_to_have', regex=True)
        # Supprimer caractères spéciaux
        s = s.str.replace(r'[^a-z0-9_\s]', '', regex=True)
        s = s.str.replace(r'\s+', '_', regex=True)
        s = s.str.strip('_')
        s = s.replace({'nan': np.nan, '': np.nan})
        df[col] = s
    return df

train_df = clean_strings(train_raw)
test_df  = clean_strings(test_raw)

print("✓ Nettoyage effectué")
print("\nValeurs 'funeral_insurance':")
print(train_df['funeral_insurance'].value_counts(dropna=False).head(6))
```

```python
# ── 3.2 Définition des types de colonnes ──────────────────────────────────
NUM_COLS = ['owner_age', 'personal_income', 'business_expenses',
            'business_turnover', 'business_age_years', 'business_age_months']
NUM_COLS = [c for c in NUM_COLS if c in train_df.columns]

# Colonnes ordinales (have_now > used_to_have > never_had)
ORDINAL_3 = ['funeral_insurance', 'medical_insurance', 'motor_vehicle_insurance',
             'has_credit_card', 'has_loan_account', 'has_internet_banking',
             'has_debit_card', 'has_mobile_money', 'uses_friends_family_savings',
             'uses_informal_lender']
ORDINAL_3 = [c for c in ORDINAL_3 if c in train_df.columns]

# Colonnes binaires (yes/no/dont_know)
BINARY_COLS = ['attitude_stable_business_environment', 'attitude_worried_shutdown',
               'compliance_income_tax', 'perception_insurance_doesnt_cover_losses',
               'perception_cannot_afford_insurance', 'current_problem_cash_flow',
               'has_cellphone', 'offers_credit_to_customers',
               'attitude_satisfied_with_achievement', 'keeps_financial_records',
               'perception_insurance_companies_dont_insure_businesses_like_yours',
               'perception_insurance_important', 'has_insurance',
               'covid_essential_service', 'attitude_more_successful_next_year',
               'problem_sourcing_money', 'marketing_word_of_mouth',
               'future_risk_theft_stock', 'motivation_make_more_money']
BINARY_COLS = [c for c in BINARY_COLS if c in train_df.columns]

# Colonnes nominales
NOMINAL_COLS = ['country', 'owner_sex']
NOMINAL_COLS = [c for c in NOMINAL_COLS if c in train_df.columns]

print(f"Numériques     : {len(NUM_COLS)} — {NUM_COLS}")
print(f"Ordinales (3niv): {len(ORDINAL_3)} — {ORDINAL_3}")
print(f"Binaires       : {len(BINARY_COLS)}")
print(f"Nominales      : {len(NOMINAL_COLS)} — {NOMINAL_COLS}")
```

```python
# ── 3.3 Encodage ordinal des features 3 niveaux ───────────────────────────
# have_now=2, used_to_have=1, never_had=0, NaN → flag séparé
ORDINAL_MAP = {
    'have_now': 2,
    'used_to_have': 1,
    'never_had': 0,
    'yes': 2,
    'no': 0,
    'dont_know': -1,
    np.nan: -2,  # codé séparément
}

def encode_ordinal(df, cols):
    df = df.copy()
    for col in cols:
        if col not in df.columns: continue
        # Flag NaN avant encodage
        df[f'{col}_isna'] = df[col].isnull().astype(int)
        # Encoder
        df[col] = df[col].map(ORDINAL_MAP).fillna(-1)  # inconnu=-1
    return df

train_df = encode_ordinal(train_df, ORDINAL_3)
test_df  = encode_ordinal(test_df,  ORDINAL_3)
print("✓ Encodage ordinal effectué")
```

```python
# ── 3.4 Encodage binaire ──────────────────────────────────────────────────
BINARY_MAP = {'yes': 1, 'no': 0, 'dont_know': -1, 'yes_sometimes': 1,
              'yes_always': 2, 'no_dont_have': 0}

def encode_binary(df, cols):
    df = df.copy()
    for col in cols:
        if col not in df.columns: continue
        df[f'{col}_isna'] = df[col].isnull().astype(int)
        # Simplification spéciale pour keeps_financial_records
        if col == 'keeps_financial_records':
            df[col] = df[col].map({'yes_always': 2, 'yes_sometimes': 1, 'no': 0}).fillna(-1)
        else:
            df[col] = df[col].map({'yes': 1, 'no': 0, 'dont_know': -1,
                                   'yes_sometimes': 1, 'yes_always': 1}).fillna(-1)
    return df

train_df = encode_binary(train_df, BINARY_COLS)
test_df  = encode_binary(test_df,  BINARY_COLS)
print("✓ Encodage binaire effectué")
```

```python
# ── 3.5 Encodage nominal (Label Encoding) ─────────────────────────────────
for col in NOMINAL_COLS:
    for df in [train_df, test_df]:
        df[f'{col}_isna'] = df[col].isnull().astype(int)

le_country = LabelEncoder()
train_df['country_enc'] = le_country.fit_transform(train_df['country'].fillna('unknown'))
test_df['country_enc']  = le_country.transform(test_df['country'].fillna('unknown'))

le_sex = LabelEncoder()
train_df['owner_sex_enc'] = le_sex.fit_transform(train_df['owner_sex'].fillna('unknown'))
test_df['owner_sex_enc']  = le_sex.transform(test_df['owner_sex'].fillna('unknown'))

print("✓ Pays encodés:", dict(zip(le_country.classes_, le_country.transform(le_country.classes_))))
```

```python
# ── 3.6 Imputation numérique ──────────────────────────────────────────────
# Flags NaN pour les numériques
for col in NUM_COLS:
    for df in [train_df, test_df]:
        df[f'{col}_isna'] = df[col].isnull().astype(int)

# Imputation médiane (par pays pour plus de précision)
for col in NUM_COLS:
    country_medians = train_df.groupby('country')[col].median()
    global_median   = train_df[col].median()
    for df in [train_df, test_df]:
        mask_na = df[col].isnull()
        if mask_na.any():
            df.loc[mask_na, col] = df.loc[mask_na, 'country'].map(country_medians).fillna(global_median)

print("✓ Imputation numérique par pays + médiane globale")
print(f"NaN restants : {train_df[NUM_COLS].isnull().sum().sum()}")
```

# 4. Feature Engineering Avancé

```python
# ── 4.1 Ratios financiers ──────────────────────────────────────────────────
EPS = 1.0  # éviter log(0)

def add_financial_features(df):
    df = df.copy()
    t = df['business_turnover']
    e = df['business_expenses']
    inc = df['personal_income']
    age_m = df['business_age_months'].clip(1)
    age_y = df['business_age_years'].clip(0.1)

    # Profit
    df['profit']        = t - e
    df['profit_margin'] = df['profit'] / (t + EPS)                     # [-∞, 1]
    df['expense_ratio'] = e / (t + EPS)                                 # coût / revenu

    # Rendement
    df['turnover_per_year']  = t / age_y
    df['income_to_turnover'] = inc / (t + EPS)
    df['surplus']            = inc - e                                  # cash disponible

    # Log-transforms (réduit l'asymétrie)
    df['log_turnover']  = np.log1p(t.clip(0))
    df['log_expenses']  = np.log1p(e.clip(0))
    df['log_income']    = np.log1p(inc.clip(0))
    df['log_profit']    = np.sign(df['profit']) * np.log1p(df['profit'].abs())

    # Ratios log
    df['log_t_over_e']  = df['log_turnover'] - df['log_expenses']
    df['log_inc_t']     = df['log_income'] - df['log_turnover']

    # Âge total en mois
    df['total_age_months'] = age_y * 12 + df['business_age_months'].fillna(0)

    return df

train_df = add_financial_features(train_df)
test_df  = add_financial_features(test_df)
print("✓ Ratios financiers ajoutés")
```

```python
# ── 4.2 Scores composites ──────────────────────────────────────────────────
def add_composite_scores(df):
    df = df.copy()

    # Score inclusion financière : somme des produits financiers (0/1/2)
    fin_cols = [c for c in ['funeral_insurance', 'medical_insurance', 'motor_vehicle_insurance',
                             'has_credit_card', 'has_loan_account', 'has_internet_banking',
                             'has_debit_card', 'has_mobile_money', 'has_insurance']
                if c in df.columns]
    df['fin_inclusion_score'] = df[fin_cols].clip(0).sum(axis=1)
    df['fin_have_now_count']  = (df[fin_cols] == 2).sum(axis=1)  # nb produits actifs
    df['fin_never_had_count'] = (df[fin_cols] == 0).sum(axis=1)  # nb jamais eu

    # Score formalité
    form_cols = [c for c in ['keeps_financial_records', 'compliance_income_tax'] if c in df.columns]
    df['formality_score'] = df[form_cols].clip(0).sum(axis=1)

    # Score risque (facteurs négatifs)
    risk_cols = [c for c in ['attitude_worried_shutdown', 'current_problem_cash_flow',
                              'problem_sourcing_money', 'uses_informal_lender',
                              'perception_cannot_afford_insurance'] if c in df.columns]
    df['risk_score'] = df[risk_cols].clip(0).sum(axis=1)

    # Score attitude positive
    att_pos_cols = [c for c in ['attitude_stable_business_environment',
                                 'attitude_more_successful_next_year',
                                 'attitude_satisfied_with_achievement'] if c in df.columns]
    df['attitude_positive_score'] = df[att_pos_cols].clip(0).sum(axis=1)

    # Score global (synthèse)
    df['global_health_proxy'] = (
        df['fin_inclusion_score'] * 2
        + df['formality_score'] * 3
        + df['attitude_positive_score']
        - df['risk_score'] * 2
    )

    return df

train_df = add_composite_scores(train_df)
test_df  = add_composite_scores(test_df)

print("Score global par classe:")
print(train_df.groupby(train_df[TARGET].map(label_map))['global_health_proxy'].mean())
```

```python
# ── 4.3 Features d'interaction ─────────────────────────────────────────────
def add_interactions(df):
    df = df.copy()

    # Interaction : formalité × finance
    df['formality_x_finance']   = df['formality_score'] * df['fin_inclusion_score']
    df['formality_x_turnover']  = df['formality_score'] * df['log_turnover']
    df['finance_x_profit']      = df['fin_inclusion_score'] * df['profit_margin'].clip(-2, 2)

    # Tranche d'âge (ordinal)
    df['age_bin'] = pd.cut(df['owner_age'],
                            bins=[0, 25, 35, 45, 55, 120], right=False,
                            labels=[0, 1, 2, 3, 4]).astype(float).fillna(2)
    df['age_x_experience']      = df['age_bin'] * np.log1p(df['business_age_years'])

    # Indicateur : profit positif
    df['is_profitable']         = (df['profit'] > 0).astype(int)
    df['is_high_margin']        = (df['profit_margin'] > 0.3).astype(int)

    # Turnover per capita (proxy)
    df['surplus_x_finance']     = df['surplus'].clip(-1e6, 1e6) * df['fin_have_now_count']

    # Nombre total de flags NaN (indicateur de qualité des données)
    na_flag_cols = [c for c in df.columns if c.endswith('_isna')]
    df['total_na_flags']        = df[na_flag_cols].sum(axis=1)

    return df

train_df = add_interactions(train_df)
test_df  = add_interactions(test_df)
print("✓ Features d'interaction ajoutées")
```

```python
# ── 4.4 Target Encoding par pays (CV-safe) ────────────────────────────────
# Pour éviter le data leakage, on calcule le target encoding uniquement sur le fold train
# Ici on calcule sur tout le train pour les prédictions test (pas de leakage)

y_train_full = train_df[TARGET].map(label_map)

def add_target_encoding(train, test, target_series, cols_to_encode, smoothing=10):
    """
    Target encoding avec smoothing pour éviter l'overfitting.
    smoothing=10 : mélange entre moyenne globale et moyenne locale.
    """
    train = train.copy()
    test  = test.copy()
    global_mean = target_series.mean()

    for col in cols_to_encode:
        if col not in train.columns: continue
        stats = pd.concat([train[col].rename('cat'), target_series.rename('target')], axis=1)
        agg   = stats.groupby('cat')['target'].agg(['mean', 'count'])
        # Smoothing: weighted average
        smooth = (agg['count'] * agg['mean'] + smoothing * global_mean) / (agg['count'] + smoothing)
        te_name = f'{col}_te'
        train[te_name] = train[col].map(smooth).fillna(global_mean)
        test[te_name]  = test[col].map(smooth).fillna(global_mean)

    return train, test

# Colonnes à target-encoder (nominales + ordinal country × classe)
TE_COLS = ['country', 'owner_sex']
# On encode aussi le pays croisé avec les produits financiers clés
train_df['country_x_funeral'] = train_df['country'].astype(str) + '_' + train_df['funeral_insurance'].astype(str)
test_df['country_x_funeral']  = test_df['country'].astype(str) + '_' + test_df['funeral_insurance'].astype(str)
TE_COLS.append('country_x_funeral')

train_df, test_df = add_target_encoding(train_df, test_df, y_train_full, TE_COLS)
print("✓ Target Encoding effectué pour :", TE_COLS)
print("Encodage pays:")
print(train_df.groupby('country')['country_te'].first())
```

---
# 5. Data Sampling & Split

```python
# ── 5.1 Construction du dataset final ─────────────────────────────────────
# Exclure les colonnes texte originales (déjà encodées)
EXCLUDE_COLS = [TARGET, ID_COL, 'country', 'owner_sex', 'country_x_funeral']

feature_cols = [c for c in train_df.columns
                if c not in EXCLUDE_COLS
                and train_df[c].dtype in ['float64', 'int64', 'int32', 'float32', 'int8', 'uint8']]

# Vérification : supprimer colonnes constantes
stds = train_df[feature_cols].std()
const_cols = stds[stds == 0].index.tolist()
if const_cols: print(f"Colonnes constantes supprimées: {const_cols}")
feature_cols = [c for c in feature_cols if c not in const_cols]

X_all  = train_df[feature_cols].copy()
y_all  = train_df[TARGET].map(label_map).values
X_test = test_df[feature_cols].copy()

# Nettoyer noms colonnes pour LightGBM
def clean_names(df):
    df.columns = (df.columns.str.replace(r'[^0-9a-zA-Z_]', '_', regex=True)
                            .str.replace(r'_+', '_', regex=True)
                            .str.strip('_'))
    return df

X_all  = clean_names(X_all)
X_test = clean_names(X_test)

# Assurer même colonnes
X_test = X_test.reindex(columns=X_all.columns, fill_value=0)

print(f"✓ Features finales : {X_all.shape[1]}")
print(f"  Train : {X_all.shape}")
print(f"  Test  : {X_test.shape}")
print(f"  Toutes numériques : {X_all.dtypes.nunique()} types")
```

```python
# ── 5.2 Split stratifié ────────────────────────────────────────────────────
X_train, X_val, y_train, y_val = train_test_split(
    X_all, y_all, test_size=0.20, random_state=SEED, stratify=y_all
)

print(f"Train : {X_train.shape} | Val : {X_val.shape}")
print(f"Classes train : {dict(zip(*np.unique(y_train, return_counts=True)))}")

# ── Sample weights : boost ×3 pour la classe High ─────────────────────────
# NE PAS utiliser class_weight='balanced' → trop agressif
# À la place : boost modéré uniquement sur High (classe 2)
HIGH_BOOST = 3.0   # multiplier les exemples High par 3
sw_train = np.where(y_train == 2, HIGH_BOOST, 1.0)
print(f"\nSample weights : Low={1.0}, Medium={1.0}, High={HIGH_BOOST}")
```

---
# 6. Modélisation — Stacking Ensemble

```python
# ── 6.1 LightGBM — Modèle principal ──────────────────────────────────────
lgbm_params = {
    'objective':        'multiclass',
    'num_class':        3,
    'metric':           'multi_logloss',
    'n_estimators':     2000,
    'learning_rate':    0.02,
    'num_leaves':       80,
    'max_depth':        -1,
    'min_child_samples': 15,
    'subsample':        0.85,
    'subsample_freq':   1,
    'colsample_bytree': 0.75,
    'reg_alpha':        0.05,
    'reg_lambda':       0.5,
    'min_split_gain':   0.01,
    # PAS de class_weight='balanced' — géré par sample_weight
    'n_jobs':           -1,
    'random_state':     SEED,
    'verbose':          -1,
}

lgbm_model = LGBMClassifier(**lgbm_params)
lgbm_model.fit(
    X_train, y_train,
    sample_weight=sw_train,
    eval_set=[(X_val, y_val)],
    callbacks=[
        lgb.early_stopping(stopping_rounds=100, verbose=False),
        lgb.log_evaluation(period=-1)
    ]
)

y_pred_lgbm = lgbm_model.predict(X_val)
print("=== LightGBM + sample_weight — Val ===")
print(classification_report(y_val, y_pred_lgbm, target_names=['Low', 'Medium', 'High']))
print(f"✓ Weighted F1 : {f1_score(y_val, y_pred_lgbm, average='weighted'):.4f}")
```

```python
# ── 6.2 XGBoost ────────────────────────────────────────────────────────────
if XGB_AVAILABLE:
    # Calculer scale_pos_weight pour XGBoost (approx pour multi-class)
    xgb_model = XGBClassifier(
        objective='multi:softprob',
        num_class=3,
        n_estimators=1500,
        learning_rate=0.03,
        max_depth=7,
        min_child_weight=5,
        subsample=0.85,
        colsample_bytree=0.75,
        colsample_bylevel=0.8,
        reg_alpha=0.05,
        reg_lambda=1.0,
        eval_metric='mlogloss',
        early_stopping_rounds=80,
        n_jobs=-1,
        random_state=SEED,
        verbosity=0
    )
    xgb_model.fit(X_train, y_train,
                  sample_weight=sw_train,
                  eval_set=[(X_val, y_val)],
                  verbose=False)
    y_pred_xgb = xgb_model.predict(X_val)
    print("=== XGBoost + sample_weight — Val ===")
    print(classification_report(y_val, y_pred_xgb, target_names=['Low', 'Medium', 'High']))
    print(f"✓ Weighted F1 : {f1_score(y_val, y_pred_xgb, average='weighted'):.4f}")
else:
    print("XGBoost non disponible")
```

```python
# ── 6.3 CatBoost (sans class_weight agressif) ─────────────────────────────
if CB_AVAILABLE:
    # class_weights modéré au lieu de 'Balanced'
    n_low    = np.sum(y_train == 0)
    n_med    = np.sum(y_train == 1)
    n_high   = np.sum(y_train == 2)
    n_total  = len(y_train)
    # Poids inversement proportionnel mais plafonné
    w_low  = min(n_total / (3 * n_low),  1.5)
    w_med  = min(n_total / (3 * n_med),  1.5)
    w_high = min(n_total / (3 * n_high), 5.0)

    cb_model = CatBoostClassifier(
        iterations=1500,
        learning_rate=0.03,
        depth=8,
        l2_leaf_reg=3,
        class_weights=[w_low, w_med, w_high],
        eval_metric='WKappa',
        early_stopping_rounds=100,
        random_seed=SEED,
        verbose=False
    )
    cb_model.fit(X_train, y_train, eval_set=(X_val, y_val))
    y_pred_cb = cb_model.predict(X_val).flatten().astype(int)
    print(f"CatBoost weights: Low={w_low:.2f}, Med={w_med:.2f}, High={w_high:.2f}")
    print("=== CatBoost (poids modérés) — Val ===")
    print(classification_report(y_val, y_pred_cb, target_names=['Low', 'Medium', 'High']))
    print(f"✓ Weighted F1 : {f1_score(y_val, y_pred_cb, average='weighted'):.4f}")
else:
    print("CatBoost non disponible")
```

```python
# ── 6.4 Random Forest ─────────────────────────────────────────────────────
rf_model = RandomForestClassifier(
    n_estimators=500,
    max_depth=20,
    min_samples_leaf=5,
    max_features='sqrt',
    class_weight={0: 1.0, 1: 1.0, 2: 4.0},  # boost High
    n_jobs=-1,
    random_state=SEED
)
rf_model.fit(X_train, y_train)
y_pred_rf = rf_model.predict(X_val)
print("=== Random Forest — Val ===")
print(classification_report(y_val, y_pred_rf, target_names=['Low', 'Medium', 'High']))
print(f"✓ Weighted F1 : {f1_score(y_val, y_pred_rf, average='weighted'):.4f}")
```

```python
# ── 6.5 Stacking — Out-Of-Fold méta-features ──────────────────────────────
# Principe : générer des OOF predictions → les utiliser comme features pour un méta-classifieur
print("Génération des OOF predictions pour stacking...")

SKF = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)

# Tableaux OOF pour le train
oof_lgbm  = np.zeros((len(X_all), 3))
oof_rf    = np.zeros((len(X_all), 3))
oof_xgb   = np.zeros((len(X_all), 3)) if XGB_AVAILABLE else None

# Prédictions test (moyenne sur les folds)
test_lgbm_folds = np.zeros((len(X_test), 3))
test_rf_folds   = np.zeros((len(X_test), 3))
test_xgb_folds  = np.zeros((len(X_test), 3)) if XGB_AVAILABLE else None

for fold, (tr_idx, oof_idx) in enumerate(SKF.split(X_all, y_all)):
    X_tr, X_oof = X_all.iloc[tr_idx], X_all.iloc[oof_idx]
    y_tr, y_oof = y_all[tr_idx], y_all[oof_idx]
    sw_tr = np.where(y_tr == 2, HIGH_BOOST, 1.0)

    # LightGBM
    m_lgbm = LGBMClassifier(**lgbm_params)
    m_lgbm.fit(X_tr, y_tr, sample_weight=sw_tr,
               eval_set=[(X_oof, y_oof)],
               callbacks=[lgb.early_stopping(100, verbose=False), lgb.log_evaluation(-1)])
    oof_lgbm[oof_idx] = m_lgbm.predict_proba(X_oof)
    test_lgbm_folds  += m_lgbm.predict_proba(X_test) / SKF.n_splits

    # Random Forest
    m_rf = RandomForestClassifier(n_estimators=300, max_depth=20, min_samples_leaf=5,
                                   max_features='sqrt', class_weight={0:1,1:1,2:4},
                                   n_jobs=-1, random_state=SEED)
    m_rf.fit(X_tr, y_tr)
    oof_rf[oof_idx] = m_rf.predict_proba(X_oof)
    test_rf_folds  += m_rf.predict_proba(X_test) / SKF.n_splits

    # XGBoost
    if XGB_AVAILABLE:
        m_xgb = XGBClassifier(objective='multi:softprob', num_class=3,
                               n_estimators=500, learning_rate=0.05,
                               max_depth=7, subsample=0.85,
                               colsample_bytree=0.75, reg_lambda=1.0,
                               eval_metric='mlogloss', early_stopping_rounds=50,
                               n_jobs=-1, random_state=SEED, verbosity=0)
        m_xgb.fit(X_tr, y_tr, sample_weight=sw_tr,
                  eval_set=[(X_oof, y_oof)], verbose=False)
        oof_xgb[oof_idx]    = m_xgb.predict_proba(X_oof)
        test_xgb_folds     += m_xgb.predict_proba(X_test) / SKF.n_splits

    f1_fold = f1_score(y_oof, np.argmax(oof_lgbm[oof_idx], axis=1), average='weighted')
    print(f"  Fold {fold+1}/5 — LGBM OOF F1: {f1_fold:.4f}")

# Score OOF global
oof_lgbm_preds = np.argmax(oof_lgbm, axis=1)
print(f"\n✓ OOF LGBM Weighted F1 : {f1_score(y_all, oof_lgbm_preds, average='weighted'):.4f}")
```

```python
# ── 6.6 Méta-classifieur (Level 2) ────────────────────────────────────────
# Construire la matrice méta : OOF probas des base models
if XGB_AVAILABLE:
    meta_train = np.hstack([oof_lgbm, oof_rf, oof_xgb])
    meta_test  = np.hstack([test_lgbm_folds, test_rf_folds, test_xgb_folds])
else:
    meta_train = np.hstack([oof_lgbm, oof_rf])
    meta_test  = np.hstack([test_lgbm_folds, test_rf_folds])

print(f"Méta-features train : {meta_train.shape}")

# Méta-classifieur : LogisticRegression (simple, évite l'overfitting)
meta_clf = LogisticRegression(
    multi_class='multinomial',
    C=1.0,
    max_iter=1000,
    random_state=SEED,
    class_weight={0: 1.0, 1: 1.0, 2: 2.5}  # léger boost High
)
meta_clf.fit(meta_train, y_all)

# Score OOF du stack
meta_preds_oof = meta_clf.predict(meta_train)
print(f"\n✓ Stack OOF Weighted F1 : {f1_score(y_all, meta_preds_oof, average='weighted'):.4f}")
print(classification_report(y_all, meta_preds_oof, target_names=['Low', 'Medium', 'High']))
```

```python
# ── 7.1 Résumé des performances validation ────────────────────────────────
models_eval = {
    'LightGBM': y_pred_lgbm,
    'Random Forest': y_pred_rf,
}
if XGB_AVAILABLE: models_eval['XGBoost'] = y_pred_xgb
if CB_AVAILABLE:  models_eval['CatBoost'] = y_pred_cb

# Soft voting sur val
val_probas = [lgbm_model.predict_proba(X_val), rf_model.predict_proba(X_val)]
if XGB_AVAILABLE: val_probas.append(xgb_model.predict_proba(X_val))
if CB_AVAILABLE:  val_probas.append(cb_model.predict_proba(X_val))
soft_vote_val = np.argmax(np.mean(val_probas, axis=0), axis=1)
models_eval['Soft Voting'] = soft_vote_val

results_df = pd.DataFrame([
    {'Modèle': name,
     'Weighted F1': f1_score(y_val, preds, average='weighted'),
     'F1 Low': f1_score(y_val, preds, average=None, labels=[0,1,2])[0],
     'F1 Medium': f1_score(y_val, preds, average=None, labels=[0,1,2])[1],
     'F1 High': f1_score(y_val, preds, average=None, labels=[0,1,2])[2]}
    for name, preds in models_eval.items()
]).sort_values('Weighted F1', ascending=False)

print(results_df.round(4).to_string(index=False))

fig, axes = plt.subplots(1, 2, figsize=(14, 4))
colors_bar = ['#2ecc71' if v > 0.88 else '#e74c3c' for v in results_df['Weighted F1']]
axes[0].barh(results_df['Modèle'][::-1], results_df['Weighted F1'][::-1],
             color=colors_bar[::-1], edgecolor='black', lw=0.7)
axes[0].axvline(0.88, color='navy', ls='--', lw=1.5, label='Baseline 0.88')
axes[0].axvline(0.92, color='green', ls='--', lw=1.5, label='Objectif 0.92')
axes[0].set_xlim(0.80, 1.0)
axes[0].set_title('Weighted F1 — Validation', fontweight='bold')
axes[0].legend(fontsize=8)
for i, v in enumerate(results_df['Weighted F1'][::-1]):
    axes[0].text(v+0.001, i, f'{v:.4f}', va='center', fontsize=8)

# F1 par classe pour le meilleur modèle
best_row = results_df.iloc[0]
axes[1].bar(['Low', 'Medium', 'High'],
            [best_row['F1 Low'], best_row['F1 Medium'], best_row['F1 High']],
            color=colors_cls, edgecolor='black', lw=0.7)
axes[1].set_ylim(0, 1.05)
axes[1].set_title(f'F1 par classe — {best_row["Modèle"]}', fontweight='bold')
axes[1].axhline(0.88, color='navy', ls='--', lw=1)
for i, v in enumerate([best_row['F1 Low'], best_row['F1 Medium'], best_row['F1 High']]):
    axes[1].text(i, v+0.01, f'{v:.3f}', ha='center', fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
# ── 7.2 Matrice de confusion — Meilleur modèle ────────────────────────────
best_model_name = results_df.iloc[0]['Modèle']
best_preds_val  = models_eval[best_model_name]

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
for ax, norm, title in zip(axes, [None, 'true'],
                            ['Counts', 'Normalisée (recall)']):
    cm = confusion_matrix(y_val, best_preds_val, normalize=norm)
    if norm == 'true': cm = cm.round(2)
    ConfusionMatrixDisplay(cm, display_labels=['Low', 'Medium', 'High']).plot(
        ax=ax, cmap='Blues', colorbar=False)
    ax.set_title(f'{best_model_name} — {title}', fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
# ── 7.3 Feature Importance — LightGBM ─────────────────────────────────────
fi = pd.DataFrame({
    'feature':    lgbm_model.feature_names_in_,
    'importance': lgbm_model.feature_importances_
}).sort_values('importance', ascending=False)

TOP_N = 30
top_fi = fi.head(TOP_N)

# Catégoriser les features
fe_kw = ['profit', 'expense_ratio', 'fin_', 'formality', 'risk_score', 'global_health',
         'log_', 'surplus', 'age_x', 'is_profit', 'is_high', 'attitude',
         'formality_x', 'finance_x', 'total_na', 'total_age', 'turnover_per',
         'income_to', 'log_t_over', 'log_inc', '_te', 'country_x']

colors_fi = ['#e74c3c' if any(kw in f for kw in fe_kw) else '#3498db' for f in top_fi['feature']]

fig, ax = plt.subplots(figsize=(12, 9))
ax.barh(top_fi['feature'][::-1], top_fi['importance'][::-1], color=colors_fi[::-1], alpha=0.85)
ax.set_title(f'Top {TOP_N} Feature Importance LightGBM (rouge = features engineerées)',
             fontweight='bold')
ax.set_xlabel('Importance (split count)')
legend_elems = [Patch(facecolor='#e74c3c', label='Feature engineerée'),
                Patch(facecolor='#3498db', label='Feature originale')]
ax.legend(handles=legend_elems)
plt.tight_layout()
plt.show()

print("\nTop 10 features :")
print(fi.head(10).to_string(index=False))
```

```python
# ── 7.4 Optimisation du seuil de décision pour High ──────────────────────
# L'idée : ajuster le seuil de classification pour High (classe 2)
# afin de maximiser le F1 weighted

lgbm_probas_val = lgbm_model.predict_proba(X_val)  # (N, 3)

best_f1 = 0
best_thresh = 0.5
thresh_results = []

for thresh_high in np.arange(0.10, 0.70, 0.02):
    # Prédire High si proba_high > thresh_high, sinon argmax parmi Low/Medium
    preds_adj = []
    for proba in lgbm_probas_val:
        if proba[2] >= thresh_high:
            preds_adj.append(2)
        else:
            preds_adj.append(np.argmax(proba[:2]))
    f1 = f1_score(y_val, preds_adj, average='weighted')
    thresh_results.append((thresh_high, f1))
    if f1 > best_f1:
        best_f1 = f1
        best_thresh = thresh_high

thresh_df = pd.DataFrame(thresh_results, columns=['threshold_high', 'weighted_f1'])

fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(thresh_df['threshold_high'], thresh_df['weighted_f1'], 'b-o', ms=4)
ax.axvline(best_thresh, color='red', ls='--', label=f'Optimal = {best_thresh:.2f}')
ax.axhline(best_f1, color='green', ls='--', label=f'Best F1 = {best_f1:.4f}')
ax.set_xlabel('Seuil de décision pour classe High')
ax.set_ylabel('Weighted F1')
ax.set_title('Optimisation du seuil de décision pour High', fontweight='bold')
ax.legend()
plt.tight_layout()
plt.show()

print(f"\nMeilleur seuil High : {best_thresh:.2f} → F1 = {best_f1:.4f}")
```

---
# 8. Entraînement Final & Soumission

```python
# ── 8.1 Sélectionner la meilleure stratégie ────────────────────────────────
# Comparer : Soft Voting vs Stack vs LGBM+threshold
scores_compare = {}

# Soft Voting
scores_compare['Soft Voting'] = f1_score(y_val, soft_vote_val, average='weighted')

# LGBM + threshold optimisé
preds_thresh = []
for proba in lgbm_probas_val:
    if proba[2] >= best_thresh:
        preds_thresh.append(2)
    else:
        preds_thresh.append(np.argmax(proba[:2]))
scores_compare['LGBM + Threshold'] = f1_score(y_val, preds_thresh, average='weighted')

# Blending pondéré : donner plus de poids au LGBM (généralement le meilleur)
if XGB_AVAILABLE and CB_AVAILABLE:
    w_lgbm, w_xgb, w_rf, w_cb = 0.40, 0.25, 0.10, 0.25
    blend_val = (w_lgbm * lgbm_model.predict_proba(X_val)
                + w_xgb * xgb_model.predict_proba(X_val)
                + w_rf  * rf_model.predict_proba(X_val)
                + w_cb  * cb_model.predict_proba(X_val))
elif XGB_AVAILABLE:
    w_lgbm, w_xgb, w_rf = 0.50, 0.30, 0.20
    blend_val = (w_lgbm * lgbm_model.predict_proba(X_val)
                + w_xgb * xgb_model.predict_proba(X_val)
                + w_rf  * rf_model.predict_proba(X_val))
else:
    blend_val = lgbm_model.predict_proba(X_val)

blend_preds_val = np.argmax(blend_val, axis=1)
scores_compare['Weighted Blending'] = f1_score(y_val, blend_preds_val, average='weighted')

print("\n=== Comparaison finale ===")
for name, score in sorted(scores_compare.items(), key=lambda x: -x[1]):
    marker = '✓ BEST' if score == max(scores_compare.values()) else ''
    print(f"  {name:25s}: {score:.4f}  {marker}")

best_strategy = max(scores_compare, key=scores_compare.get)
print(f"\n→ Stratégie retenue : {best_strategy}")
```

```python
# ── 8.2 Entraîner sur TOUT le train et prédire le test ────────────────────
print("Entraînement final sur l'intégralité du train...")

sw_all = np.where(y_all == 2, HIGH_BOOST, 1.0)

# LightGBM final (utiliser le n_estimators du best_iteration)
n_est_final = lgbm_model.best_iteration_ + 50 if hasattr(lgbm_model, 'best_iteration_') else 1200
lgbm_params_final = dict(lgbm_params)
lgbm_params_final['n_estimators'] = n_est_final

lgbm_final = LGBMClassifier(**lgbm_params_final)
lgbm_final.fit(X_all, y_all, sample_weight=sw_all,
               callbacks=[lgb.log_evaluation(-1)])

# RF final
rf_final = RandomForestClassifier(
    n_estimators=500, max_depth=20, min_samples_leaf=5,
    max_features='sqrt', class_weight={0:1,1:1,2:4}, n_jobs=-1, random_state=SEED)
rf_final.fit(X_all, y_all)

# XGBoost final
if XGB_AVAILABLE:
    n_xgb_final = xgb_model.best_iteration + 50 if hasattr(xgb_model, 'best_iteration') else 600
    xgb_final = XGBClassifier(
        objective='multi:softprob', num_class=3,
        n_estimators=n_xgb_final, learning_rate=0.03,
        max_depth=7, subsample=0.85, colsample_bytree=0.75,
        reg_lambda=1.0, n_jobs=-1, random_state=SEED, verbosity=0)
    xgb_final.fit(X_all, y_all, sample_weight=sw_all, verbose=False)

# CatBoost final
if CB_AVAILABLE:
    n_cb_final = cb_model.best_iteration_ + 50 if hasattr(cb_model, 'best_iteration_') else 600
    cb_final = CatBoostClassifier(
        iterations=n_cb_final, learning_rate=0.03, depth=8, l2_leaf_reg=3,
        class_weights=[w_low, w_med, w_high], random_seed=SEED, verbose=False)
    cb_final.fit(X_all, y_all)

print("✓ Entraînement final terminé")
```

```python
# ── 8.3 Prédictions test avec la meilleure stratégie ──────────────────────
print(f"Génération des prédictions avec : {best_strategy}")

# Calcul des probabilités test
test_proba_lgbm = lgbm_final.predict_proba(X_test)
test_proba_rf   = rf_final.predict_proba(X_test)

if XGB_AVAILABLE and CB_AVAILABLE:
    test_proba_xgb = xgb_final.predict_proba(X_test)
    test_proba_cb  = cb_final.predict_proba(X_test)
    blend_test = (w_lgbm * test_proba_lgbm
                 + w_xgb * test_proba_xgb
                 + w_rf  * test_proba_rf
                 + w_cb  * test_proba_cb)
elif XGB_AVAILABLE:
    test_proba_xgb = xgb_final.predict_proba(X_test)
    blend_test = (w_lgbm * test_proba_lgbm
                 + w_xgb * test_proba_xgb
                 + w_rf  * test_proba_rf)
else:
    blend_test = test_proba_lgbm

# Appliquer le seuil optimisé pour High
final_preds_enc = []
for proba in blend_test:
    if proba[2] >= best_thresh:
        final_preds_enc.append(2)
    else:
        final_preds_enc.append(np.argmax(proba[:2]))

final_preds = [inv_label_map[p] for p in final_preds_enc]

print("\nDistribution des prédictions test:")
uniq, cnts = np.unique(final_preds, return_counts=True)
for u, c in zip(uniq, cnts):
    print(f"  {u:8s}: {c:4d} ({100*c/len(final_preds):.1f}%)")

# Sauvegarder
submission = pd.DataFrame({'ID': test_raw['ID'].values, 'Target': final_preds})
submission.to_csv('submission.csv', index=False)
print(f"\n✓ submission.csv sauvegardé ({len(submission)} lignes)")
submission.head()
```
