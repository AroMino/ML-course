# 0. Problem Framing

> **Tâche :** Classification supervisée 3 classes (Low / Medium / High) sur 9618 entreprises africaines  
> **Métrique :** Weighted F1-score  
> **Déséquilibre :** Low 65% / Medium 30% / High 5% — la classe High est le principal enjeu  
> **Baseline à battre :** 0.8875 → Objectif : **0.92+**
>
> **Stratégie :**  
> 1. Target Encoding safe (CV) pour les catégorielles → réduit le bruit vs OHE  
> 2. Feature engineering financier avancé + interactions pays  
> 3. LightGBM sans class_weight + **sample_weight boost pour High**  
> 4. Stacking OOF (LightGBM + XGBoost + RandomForest → LogReg méta)  
> 5. Blending final avec poids optimisés

# 3. Preprocessing Pipeline

```python
# ── 3.1 Copie & nettoyage des valeurs catégorielles ────────────────────────
def clean_strings(df):
    df = df.copy()
    for col in df.select_dtypes('object').columns:
        if col in [ID_COL]: continue
        s = df[col].astype(str).str.lower().str.strip()
        # Standardiser les variantes de "don't know"
        s = s.str.replace(r"don[?'\u2019\u0301]t know(?: or n/a)?", "dont_know", regex=True)
        # Standardiser "have now" → 'have_now'
        s = s.str.replace('have now', 'have_now', regex=False)
        # Standardiser "never had" → 'never_had'
        s = s.str.replace('never had', 'never_had', regex=False)
        # Standardiser "used to have"
        s = s.str.replace(r'used to have.*', 'used_to_have', regex=True)
        # Supprimer caractères spéciaux
        s = s.str.replace(r'[^a-z0-9_\s]', '', regex=True)
        s = s.str.replace(r'\s+', '_', regex=True)
        s = s.str.strip('_')
        s = s.replace({'nan': np.nan, '': np.nan})
        df[col] = s
    return df

train_df = clean_strings(train_raw)
test_df  = clean_strings(test_raw)

print("✓ Nettoyage effectué")
print("\nValeurs 'funeral_insurance':")
print(train_df['funeral_insurance'].value_counts(dropna=False).head(6))
```

```python
# ── 3.2 Définition des types de colonnes ──────────────────────────────────
NUM_COLS = ['owner_age', 'personal_income', 'business_expenses',
            'business_turnover', 'business_age_years', 'business_age_months']
NUM_COLS = [c for c in NUM_COLS if c in train_df.columns]

# Colonnes ordinales (have_now > used_to_have > never_had)
ORDINAL_3 = ['funeral_insurance', 'medical_insurance', 'motor_vehicle_insurance',
             'has_credit_card', 'has_loan_account', 'has_internet_banking',
             'has_debit_card', 'has_mobile_money', 'uses_friends_family_savings',
             'uses_informal_lender']
ORDINAL_3 = [c for c in ORDINAL_3 if c in train_df.columns]

# Colonnes binaires (yes/no/dont_know)
BINARY_COLS = ['attitude_stable_business_environment', 'attitude_worried_shutdown',
               'compliance_income_tax', 'perception_insurance_doesnt_cover_losses',
               'perception_cannot_afford_insurance', 'current_problem_cash_flow',
               'has_cellphone', 'offers_credit_to_customers',
               'attitude_satisfied_with_achievement', 'keeps_financial_records',
               'perception_insurance_companies_dont_insure_businesses_like_yours',
               'perception_insurance_important', 'has_insurance',
               'covid_essential_service', 'attitude_more_successful_next_year',
               'problem_sourcing_money', 'marketing_word_of_mouth',
               'future_risk_theft_stock', 'motivation_make_more_money']
BINARY_COLS = [c for c in BINARY_COLS if c in train_df.columns]

# Colonnes nominales
NOMINAL_COLS = ['country', 'owner_sex']
NOMINAL_COLS = [c for c in NOMINAL_COLS if c in train_df.columns]

print(f"Numériques     : {len(NUM_COLS)} — {NUM_COLS}")
print(f"Ordinales (3niv): {len(ORDINAL_3)} — {ORDINAL_3}")
print(f"Binaires       : {len(BINARY_COLS)}")
print(f"Nominales      : {len(NOMINAL_COLS)} — {NOMINAL_COLS}")
```

```python
# ── 3.3 Encodage ordinal des features 3 niveaux ───────────────────────────
# have_now=2, used_to_have=1, never_had=0, NaN → flag séparé
ORDINAL_MAP = {
    'have_now': 2,
    'used_to_have': 1,
    'never_had': 0,
    'yes': 2,
    'no': 0,
    'dont_know': -1,
    np.nan: -2,  # codé séparément
}

def encode_ordinal(df, cols):
    df = df.copy()
    for col in cols:
        if col not in df.columns: continue
        # Flag NaN avant encodage
        df[f'{col}_isna'] = df[col].isnull().astype(int)
        # Encoder
        df[col] = df[col].map(ORDINAL_MAP).fillna(-1)  # inconnu=-1
    return df

train_df = encode_ordinal(train_df, ORDINAL_3)
test_df  = encode_ordinal(test_df,  ORDINAL_3)
print("✓ Encodage ordinal effectué")
```

```python
# ── 3.4 Encodage binaire ──────────────────────────────────────────────────
BINARY_MAP = {'yes': 1, 'no': 0, 'dont_know': -1, 'yes_sometimes': 1,
              'yes_always': 2, 'no_dont_have': 0}

def encode_binary(df, cols):
    df = df.copy()
    for col in cols:
        if col not in df.columns: continue
        df[f'{col}_isna'] = df[col].isnull().astype(int)
        # Simplification spéciale pour keeps_financial_records
        if col == 'keeps_financial_records':
            df[col] = df[col].map({'yes_always': 2, 'yes_sometimes': 1, 'no': 0}).fillna(-1)
        else:
            df[col] = df[col].map({'yes': 1, 'no': 0, 'dont_know': -1,
                                   'yes_sometimes': 1, 'yes_always': 1}).fillna(-1)
    return df

train_df = encode_binary(train_df, BINARY_COLS)
test_df  = encode_binary(test_df,  BINARY_COLS)
print("✓ Encodage binaire effectué")
```

```python
# ── 3.5 Encodage nominal (Label Encoding) ─────────────────────────────────
for col in NOMINAL_COLS:
    for df in [train_df, test_df]:
        df[f'{col}_isna'] = df[col].isnull().astype(int)

le_country = LabelEncoder()
train_df['country_enc'] = le_country.fit_transform(train_df['country'].fillna('unknown'))
test_df['country_enc']  = le_country.transform(test_df['country'].fillna('unknown'))

le_sex = LabelEncoder()
train_df['owner_sex_enc'] = le_sex.fit_transform(train_df['owner_sex'].fillna('unknown'))
test_df['owner_sex_enc']  = le_sex.transform(test_df['owner_sex'].fillna('unknown'))

print("✓ Pays encodés:", dict(zip(le_country.classes_, le_country.transform(le_country.classes_))))
```

```python
# ── 3.6 Imputation numérique ──────────────────────────────────────────────
# Flags NaN pour les numériques
for col in NUM_COLS:
    for df in [train_df, test_df]:
        df[f'{col}_isna'] = df[col].isnull().astype(int)

# Imputation médiane (par pays pour plus de précision)
for col in NUM_COLS:
    country_medians = train_df.groupby('country')[col].median()
    global_median   = train_df[col].median()
    for df in [train_df, test_df]:
        mask_na = df[col].isnull()
        if mask_na.any():
            df.loc[mask_na, col] = df.loc[mask_na, 'country'].map(country_medians).fillna(global_median)

print("✓ Imputation numérique par pays + médiane globale")
print(f"NaN restants : {train_df[NUM_COLS].isnull().sum().sum()}")
```

# 4. Feature Engineering Avancé

```python
# ── 4.1 Ratios financiers améliorés ───────────────────────────────────────
EPS = 1.0

def add_financial_features(df):
    df = df.copy()
    t     = df['business_turnover'].fillna(0)
    e     = df['business_expenses'].fillna(0)
    inc   = df['personal_income'].fillna(0)
    age_m = df['business_age_months'].fillna(0).clip(1)
    age_y = df['business_age_years'].fillna(0).clip(0.1)

    # Profit & marges
    df['profit']        = t - e
    df['profit_margin'] = (df['profit'] / (t + EPS)).clip(-10, 10)
    df['expense_ratio'] = (e / (t + EPS)).clip(0, 10)
    df['op_leverage']   = ((t - e) / (inc + EPS)).clip(-5, 5)

    # Rendement & efficacité
    df['turnover_per_year']    = (t / age_y).clip(0, 1e9)
    df['income_to_turnover']   = (inc / (t + EPS)).clip(-10, 10)
    df['surplus']              = inc - e
    df['surplus_ratio']        = (df['surplus'] / (inc + EPS)).clip(-5, 5)
    df['income_per_age_month'] = (inc / age_m).clip(0, 1e9)

    # Log-transforms
    df['log_turnover'] = np.log1p(t.clip(0))
    df['log_expenses'] = np.log1p(e.clip(0))
    df['log_income']   = np.log1p(inc.clip(0))
    df['log_profit']   = np.sign(df['profit']) * np.log1p(df['profit'].abs())
    df['log_surplus']  = np.sign(df['surplus']) * np.log1p(df['surplus'].abs())

    # Ratios log
    df['log_t_over_e'] = df['log_turnover'] - df['log_expenses']
    df['log_inc_t']    = df['log_income'] - df['log_turnover']
    df['log_inc_e']    = df['log_income'] - df['log_expenses']

    # Âge
    df['total_age_months']     = age_y * 12 + age_m
    df['log_business_age']     = np.log1p(df['total_age_months'])
    df['revenue_per_age']      = (t / (df['total_age_months'] + 1)).clip(0, 1e9)
    df['log_revenue_per_age']  = np.log1p(df['revenue_per_age'].clip(0))

    # Indicateurs binaires
    df['is_profitable']   = (df['profit'] > 0).astype(int)
    df['is_high_margin']  = (df['profit_margin'] > 0.3).astype(int)

    # Nettoyer inf/-inf
    df.replace([np.inf, -np.inf], np.nan, inplace=True)
    return df

train_df = add_financial_features(train_df)
test_df  = add_financial_features(test_df)
print("✓ Ratios financiers ajoutés")
```

```python
# ── 4.2 Scores composites ──────────────────────────────────────────────────
def add_composite_scores(df):
    df = df.copy()

    # Score inclusion financière : somme des produits financiers (0/1/2)
    fin_cols = [c for c in ['funeral_insurance', 'medical_insurance', 'motor_vehicle_insurance',
                             'has_credit_card', 'has_loan_account', 'has_internet_banking',
                             'has_debit_card', 'has_mobile_money', 'has_insurance']
                if c in df.columns]
    df['fin_inclusion_score'] = df[fin_cols].clip(0).sum(axis=1)
    df['fin_have_now_count']  = (df[fin_cols] == 2).sum(axis=1)  # nb produits actifs
    df['fin_never_had_count'] = (df[fin_cols] == 0).sum(axis=1)  # nb jamais eu

    # Score formalité
    form_cols = [c for c in ['keeps_financial_records', 'compliance_income_tax'] if c in df.columns]
    df['formality_score'] = df[form_cols].clip(0).sum(axis=1)

    # Score risque (facteurs négatifs)
    risk_cols = [c for c in ['attitude_worried_shutdown', 'current_problem_cash_flow',
                              'problem_sourcing_money', 'uses_informal_lender',
                              'perception_cannot_afford_insurance'] if c in df.columns]
    df['risk_score'] = df[risk_cols].clip(0).sum(axis=1)

    # Score attitude positive
    att_pos_cols = [c for c in ['attitude_stable_business_environment',
                                 'attitude_more_successful_next_year',
                                 'attitude_satisfied_with_achievement'] if c in df.columns]
    df['attitude_positive_score'] = df[att_pos_cols].clip(0).sum(axis=1)

    # Score global (synthèse)
    df['global_health_proxy'] = (
        df['fin_inclusion_score'] * 2
        + df['formality_score'] * 3
        + df['attitude_positive_score']
        - df['risk_score'] * 2
    )

    return df

train_df = add_composite_scores(train_df)
test_df  = add_composite_scores(test_df)

print("Score global par classe:")
print(train_df.groupby(train_df[TARGET].map(label_map))['global_health_proxy'].mean())
```

```python
# ── 4.3 Features d'interaction ─────────────────────────────────────────────
def add_interactions(df):
    df = df.copy()

    # Interaction formalité × finance
    df['formality_x_finance']       = df['formality_score'] * df['fin_inclusion_score']
    df['formality_x_turnover']      = df['formality_score'] * df['log_turnover']
    df['finance_x_profit']          = df['fin_inclusion_score'] * df['profit_margin'].clip(-2, 2)
    df['formality_x_profit_margin'] = df['formality_score'] * df['profit_margin'].clip(-2, 2)

    # Tranche d'âge
    df['age_bin'] = pd.cut(df['owner_age'],
                            bins=[0, 25, 35, 45, 55, 120], right=False,
                            labels=[0, 1, 2, 3, 4]).astype(float).fillna(2)
    df['age_x_experience'] = df['age_bin'] * np.log1p(df['business_age_years'])
    df['age_x_income']     = df['age_bin'] * df['log_income']

    # Produits financiers × autres scores
    df['surplus_x_finance']       = df['surplus'].clip(-1e6, 1e6) * df['fin_have_now_count']
    df['finance_x_formality']     = df['fin_have_now_count'] * df['formality_score']
    df['global_x_age']            = df['global_health_proxy'] * df['age_bin']
    df['global_health_sq']        = df['global_health_proxy'] ** 2

    # Interactions comportementales
    df['attitude_x_formality']    = df['attitude_positive_score'] * df['formality_score']
    df['risk_x_finance']          = df['risk_score'] * df['fin_inclusion_score']
    df['compliant_and_profitable'] = (
        (df.get('compliance_income_tax', pd.Series(0, index=df.index)) > 0) &
        (df['is_profitable'] == 1)
    ).astype(int)

    # Turnover × inclusion financière
    df['log_turnover_x_finance']  = df['log_turnover'] * df['fin_have_now_count']
    df['log_income_x_formality']  = df['log_income'] * df['formality_score']

    # Score d'efficacité composite
    df['efficiency_score'] = (
        df['profit_margin'].clip(0, 1) * 3
        + df['formality_score']
        + df['fin_inclusion_score'] * 0.5
        - df['risk_score']
    )

    # Total NaN flags
    na_flag_cols = [c for c in df.columns if c.endswith('_isna')]
    df['total_na_flags']  = df[na_flag_cols].sum(axis=1)
    df['nan_ratio']       = df['total_na_flags'] / (len(na_flag_cols) + 1)

    return df

train_df = add_interactions(train_df)
test_df  = add_interactions(test_df)
print("✓ Features d'interaction ajoutées")
```

```python
# ── 4.4 Target Encoding CV-safe (K-Fold anti-leakage) ─────────────────────
# IMPORTANT : on prend la target depuis train_RAW (index propre 0..N)
# train_df peut avoir un index désaligné après les transformations précédentes
y_train_full = train_raw[TARGET].map(label_map).reset_index(drop=True)
assert not y_train_full.isnull().any(), f"y_train_full contient {y_train_full.isnull().sum()} NaN !"
print(f"y_train_full : {len(y_train_full)} lignes, classes {y_train_full.value_counts().to_dict()}")

def kfold_target_encoding(train, test, target_series, cols_to_encode,
                            n_splits=5, smoothing=10, seed=42):
    """Target encoding CV-safe : chaque ligne encodée hors de son fold."""
    train = train.copy().reset_index(drop=True)
    test  = test.copy().reset_index(drop=True)

    # Aligner target sur le train (même longueur, index 0..N)
    y_arr = target_series.reset_index(drop=True).values.astype(float)
    assert len(y_arr) == len(train), f"Taille mismatch : y={len(y_arr)}, train={len(train)}"
    assert not np.isnan(y_arr).any(), "y_arr contient des NaN !"

    skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=seed)
    global_mean = float(y_arr.mean())

    for col in cols_to_encode:
        if col not in train.columns:
            continue
        te_name = f'{col}_te'
        train[te_name] = global_mean
        test_te_acc = np.zeros(len(test))

        for fold, (tr_idx, oof_idx) in enumerate(skf.split(np.arange(len(train)), y_arr)):
            fold_stats = pd.DataFrame({
                'cat':    train.iloc[tr_idx][col].values,
                'target': y_arr[tr_idx]
            }).groupby('cat')['target'].agg(['mean', 'count'])

            smooth = (fold_stats['count'] * fold_stats['mean'] + smoothing * global_mean) / \
                     (fold_stats['count'] + smoothing)

            train.loc[oof_idx, te_name] = \
                train.iloc[oof_idx][col].map(smooth).fillna(global_mean).values

            test_te_acc += test[col].map(smooth).fillna(global_mean).values

        test[te_name] = test_te_acc / n_splits

    return train, test

# Construire les colonnes d'interaction à encoder
for col in ['funeral_insurance', 'medical_insurance', 'has_insurance',
            'keeps_financial_records', 'compliance_income_tax',
            'has_credit_card', 'has_loan_account']:
    if col in train_df.columns:
        key = f'country_x_{col}'
        train_df[key] = train_df['country'].astype(str) + '_' + train_df[col].astype(str)
        test_df[key]  = test_df['country'].astype(str) + '_' + test_df[col].astype(str)

train_df['sex_x_country'] = train_df['owner_sex'].astype(str) + '_' + train_df['country'].astype(str)
test_df['sex_x_country']  = test_df['owner_sex'].astype(str) + '_' + test_df['country'].astype(str)

TE_COLS = [
    'country', 'owner_sex', 'sex_x_country',
    'country_x_funeral_insurance', 'country_x_medical_insurance',
    'country_x_has_insurance', 'country_x_keeps_financial_records',
    'country_x_compliance_income_tax', 'country_x_has_credit_card',
    'country_x_has_loan_account',
]
TE_COLS = [c for c in TE_COLS if c in train_df.columns]

train_df, test_df = kfold_target_encoding(train_df, test_df, y_train_full, TE_COLS)
print("✓ CV-safe Target Encoding effectué pour :", TE_COLS)
print("\nCorrélation country_te vs target:",
      round(train_df['country_te'].corr(y_train_full), 4))
```

```python
# ── 4.5 Features de rang & position relative au sein du pays ─────────────
# La position RELATIVE dans son pays (rang percentile) est plus informative
# que la valeur absolue (économies très différentes selon les pays).

def add_country_rank_features(train, test):
    train = train.copy()
    test  = test.copy()

    rank_cols = ['personal_income', 'business_turnover', 'business_expenses',
                 'profit', 'profit_margin', 'log_income', 'log_turnover',
                 'fin_inclusion_score', 'global_health_proxy', 'surplus',
                 'formality_score', 'total_age_months', 'efficiency_score']
    rank_cols = [c for c in rank_cols if c in train.columns]

    for col in rank_cols:
        stats = train.groupby('country')[col].agg(['mean', 'std', 'median',
                                                    lambda x: x.quantile(0.25),
                                                    lambda x: x.quantile(0.75)])
        stats.columns = ['mean', 'std', 'median', 'q25', 'q75']
        stats['std']  = stats['std'].replace(0, 1).fillna(1)
        stats['iqr']  = (stats['q75'] - stats['q25']).replace(0, 1).fillna(1)

        # Z-score within country
        for df in [train, test]:
            mu  = df['country'].map(stats['mean']).fillna(train[col].mean())
            sig = df['country'].map(stats['std']).fillna(train[col].std() + 1e-9)
            df[f'{col}_z_country'] = ((df[col] - mu) / sig).clip(-5, 5)

        # Percentile rank within country (train uniquement → pas de leakage)
        train[f'{col}_pct_country'] = train.groupby('country')[col].rank(pct=True)

        # Pour le test : approximation par IQR
        med = test['country'].map(stats['median']).fillna(train[col].median())
        iqr = test['country'].map(stats['iqr']).fillna(1)
        test[f'{col}_pct_country'] = ((test[col] - med) / iqr).clip(-3, 3)

        # Différence vs médiane du pays
        train[f'{col}_vs_med_country'] = train[col] - train['country'].map(stats['median']).fillna(0)
        test[f'{col}_vs_med_country']  = test[col]  - test['country'].map(stats['median']).fillna(0)

    train.replace([np.inf, -np.inf], np.nan, inplace=True)
    test.replace([np.inf, -np.inf], np.nan, inplace=True)

    n_new = sum(1 for c in train.columns if any(s in c for s in ['_z_country', '_pct_country', '_vs_med_country']))
    print(f"  → {n_new} rank/zscore features ajoutées")
    return train, test

train_df, test_df = add_country_rank_features(train_df, test_df)
print("✓ Features de rang within-country ajoutées")
```

```python
# ── 4.6 Patterns NaN & aggrégations contextuelles pays ───────────────────

def add_nan_and_agg_features(train, test):
    train = train.copy()
    test  = test.copy()

    # Pattern NaN : nombre de champs manquants dans les colonnes financières clés
    key_nan_cols = [c for c in ['funeral_insurance_isna', 'medical_insurance_isna',
                                 'has_credit_card_isna', 'has_loan_account_isna',
                                 'uses_informal_lender_isna', 'motivation_make_more_money_isna']
                    if c in train.columns]
    if key_nan_cols:
        for df in [train, test]:
            df['fin_nan_count']    = df[key_nan_cols].sum(axis=1)
            df['is_data_complete'] = (df[key_nan_cols].sum(axis=1) == 0).astype(int)

    # Aggrégations pays (contexte macro)
    agg_cols = ['personal_income', 'business_turnover', 'profit_margin',
                'fin_inclusion_score', 'global_health_proxy', 'formality_score']
    agg_cols = [c for c in agg_cols if c in train.columns]

    country_aggs = train.groupby('country')[agg_cols].agg(['mean', 'std'])
    country_aggs.columns = [f'country_{a}_{b}' for a, b in country_aggs.columns]
    country_aggs['std'] = country_aggs.filter(like='_std').replace(0, np.nan).fillna(1)

    for df in [train, test]:
        for col in country_aggs.columns:
            df[col] = df['country'].map(country_aggs[col]).fillna(country_aggs[col].mean())

    # Richesse macro du pays
    country_health = train.groupby('country')['global_health_proxy'].mean()
    for df in [train, test]:
        df['country_health_mean'] = df['country'].map(country_health).fillna(
            train['global_health_proxy'].mean())

    train.replace([np.inf, -np.inf], np.nan, inplace=True)
    test.replace([np.inf, -np.inf], np.nan, inplace=True)
    return train, test

train_df, test_df = add_nan_and_agg_features(train_df, test_df)
print("✓ Patterns NaN & aggrégations pays ajoutés")
```

# 5. Data Sampling & Split

```python
# ── 5.1 Construction du dataset final ─────────────────────────────────────
y_all = train_raw[TARGET].map(label_map).values
assert not pd.isnull(y_all).any(), "y_all contient des NaN !"

# Colonnes à exclure (texte non encodé)
EXCLUDE_COLS = set([TARGET, ID_COL, 'country', 'owner_sex', 'sex_x_country'] +
                   [c for c in train_df.columns
                    if (c.startswith('country_x_') and not c.endswith('_te'))
                    or c == 'country_x_funeral'])

feature_cols = [c for c in train_df.columns
                if c not in EXCLUDE_COLS
                and train_df[c].dtype in ['float64', 'int64', 'int32', 'float32', 'int8', 'uint8']]

stds = train_df[feature_cols].std()
const_cols = stds[stds == 0].index.tolist()
if const_cols:
    print(f"Colonnes constantes supprimées: {const_cols}")
feature_cols = [c for c in feature_cols if c not in const_cols]

X_all  = train_df[feature_cols].copy()
X_test = test_df[feature_cols].copy()

def clean_names(df):
    df.columns = (df.columns.str.replace(r'[^0-9a-zA-Z_]', '_', regex=True)
                            .str.replace(r'_+', '_', regex=True)
                            .str.strip('_'))
    return df

X_all  = clean_names(X_all)
X_test = clean_names(X_test)
X_test = X_test.reindex(columns=X_all.columns, fill_value=0)

# Remplacer inf/-inf AVANT fillna
X_all.replace([np.inf, -np.inf], np.nan, inplace=True)
X_test.replace([np.inf, -np.inf], np.nan, inplace=True)

# Remplir NaN résiduels par médiane train
for col in X_all.columns:
    if X_all[col].isnull().any() or X_test[col].isnull().any():
        med = X_all[col].median()
        med = 0 if pd.isnull(med) else med
        X_all[col]  = X_all[col].fillna(med)
        X_test[col] = X_test[col].fillna(med)

assert not X_all.isnull().any().any(),  "X_all contient encore des NaN !"
assert not X_test.isnull().any().any(), "X_test contient encore des NaN !"
assert not pd.isnull(y_all).any(),      "y_all contient des NaN !"

print(f"✓ Features finales : {X_all.shape[1]}")
print(f"  Train : {X_all.shape}  |  Test : {X_test.shape}")
print(f"  NaN dans X_all : {X_all.isnull().sum().sum()}  (doit être 0)")
```

```python
# ── 5.2 Split stratifié ────────────────────────────────────────────────────
X_train, X_val, y_train, y_val = train_test_split(
    X_all, y_all, test_size=0.20, random_state=SEED, stratify=y_all
)

print(f"Train : {X_train.shape} | Val : {X_val.shape}")
print(f"Classes train : {dict(zip(*np.unique(y_train, return_counts=True)))}")

HIGH_BOOST = 4.0   # augmenté : 3 → 4 (classe High très minoritaire)
sw_train = np.where(y_train == 2, HIGH_BOOST, 1.0)
print(f"\nSample weights : Low={1.0}, Medium={1.0}, High={HIGH_BOOST}")
```

```python
# ── 5.3 SMOTE — Rééquilibrage de la classe High ───────────────────────────
# High = ~5% → BorderlineSMOTE génère des exemples synthétiques sur la frontière
# IMPORTANT : appliqué uniquement sur X_train, JAMAIS sur X_all ni X_test

if SMOTE_AVAILABLE:
    print("Classes avant SMOTE :", dict(zip(*np.unique(y_train, return_counts=True))))

    smote = BorderlineSMOTE(
        sampling_strategy={2: int(np.sum(y_train == 1) * 0.4)},
        k_neighbors=5,
        random_state=SEED
    )
    X_train_sm, y_train_sm = smote.fit_resample(X_train, y_train)
    sw_train_sm = np.where(y_train_sm == 2, HIGH_BOOST * 0.8, 1.0)

    print("Classes après SMOTE  :", dict(zip(*np.unique(y_train_sm, return_counts=True))))
    print(f"Exemples High : {np.sum(y_train == 2)} → {np.sum(y_train_sm == 2)}")
else:
    X_train_sm, y_train_sm, sw_train_sm = X_train, y_train, sw_train
    print("SMOTE non disponible — pip install imbalanced-learn")
```

# 6. Modélisation — Stacking Ensemble

```python
# ── 6.1 LightGBM — Hyperparamètres affinés ───────────────────────────────
lgbm_params = {
    'objective':          'multiclass',
    'num_class':          3,
    'metric':             'multi_logloss',
    'n_estimators':       3000,
    'learning_rate':      0.015,
    'num_leaves':         127,
    'max_depth':          -1,
    'min_child_samples':  10,
    'min_child_weight':   1e-3,
    'subsample':          0.80,
    'subsample_freq':     1,
    'colsample_bytree':   0.70,
    'reg_alpha':          0.1,
    'reg_lambda':         1.0,
    'min_split_gain':     0.005,
    'max_bin':            255,
    'path_smooth':        0.1,
    'n_jobs':             -1,
    'random_state':       SEED,
    'verbose':            -1,
}

# Utiliser SMOTE si disponible
X_tr_fit = X_train_sm if SMOTE_AVAILABLE else X_train
y_tr_fit  = y_train_sm if SMOTE_AVAILABLE else y_train
sw_fit    = sw_train_sm if SMOTE_AVAILABLE else sw_train

lgbm_model = LGBMClassifier(**lgbm_params)
lgbm_model.fit(
    X_tr_fit, y_tr_fit,
    sample_weight=sw_fit,
    eval_set=[(X_val, y_val)],
    callbacks=[
        lgb.early_stopping(stopping_rounds=150, verbose=False),
        lgb.log_evaluation(period=-1)
    ]
)

y_pred_lgbm = lgbm_model.predict(X_val)
print("=== LightGBM v4 + SMOTE + sample_weight — Val ===")
print(classification_report(y_val, y_pred_lgbm, target_names=['Low', 'Medium', 'High']))
print(f"✓ Weighted F1 : {f1_score(y_val, y_pred_lgbm, average='weighted'):.4f}")
```

```python
# ── 6.2 XGBoost ────────────────────────────────────────────────────────────
if XGB_AVAILABLE:
    # Calculer scale_pos_weight pour XGBoost (approx pour multi-class)
    xgb_model = XGBClassifier(
        objective='multi:softprob',
        num_class=3,
        n_estimators=1500,
        learning_rate=0.03,
        max_depth=7,
        min_child_weight=5,
        subsample=0.85,
        colsample_bytree=0.75,
        colsample_bylevel=0.8,
        reg_alpha=0.05,
        reg_lambda=1.0,
        eval_metric='mlogloss',
        early_stopping_rounds=80,
        n_jobs=-1,
        random_state=SEED,
        verbosity=0
    )
    xgb_model.fit(X_train, y_train,
                  sample_weight=sw_train,
                  eval_set=[(X_val, y_val)],
                  verbose=False)
    y_pred_xgb = xgb_model.predict(X_val)
    print("=== XGBoost + sample_weight — Val ===")
    print(classification_report(y_val, y_pred_xgb, target_names=['Low', 'Medium', 'High']))
    print(f"✓ Weighted F1 : {f1_score(y_val, y_pred_xgb, average='weighted'):.4f}")
else:
    print("XGBoost non disponible")
```

```python
# ── 6.3 CatBoost (sans class_weight agressif) ─────────────────────────────
if CB_AVAILABLE:
    # class_weights modéré au lieu de 'Balanced'
    n_low    = np.sum(y_train == 0)
    n_med    = np.sum(y_train == 1)
    n_high   = np.sum(y_train == 2)
    n_total  = len(y_train)
    # Poids inversement proportionnel mais plafonné
    w_low  = min(n_total / (3 * n_low),  1.5)
    w_med  = min(n_total / (3 * n_med),  1.5)
    w_high = min(n_total / (3 * n_high), 5.0)

    cb_model = CatBoostClassifier(
        iterations=1500,
        learning_rate=0.03,
        depth=8,
        l2_leaf_reg=3,
        class_weights=[w_low, w_med, w_high],
        eval_metric='WKappa',
        early_stopping_rounds=100,
        random_seed=SEED,
        verbose=False
    )
    cb_model.fit(X_train, y_train, eval_set=(X_val, y_val))
    y_pred_cb = cb_model.predict(X_val).flatten().astype(int)
    print(f"CatBoost weights: Low={w_low:.2f}, Med={w_med:.2f}, High={w_high:.2f}")
    print("=== CatBoost (poids modérés) — Val ===")
    print(classification_report(y_val, y_pred_cb, target_names=['Low', 'Medium', 'High']))
    print(f"✓ Weighted F1 : {f1_score(y_val, y_pred_cb, average='weighted'):.4f}")
else:
    print("CatBoost non disponible")
```

```python
# ── 6.4 Random Forest & Extra Trees ─────────────────────────────────────
rf_model = RandomForestClassifier(
    n_estimators=600,
    max_depth=20,
    min_samples_leaf=4,
    max_features='sqrt',
    class_weight={0: 1.0, 1: 1.0, 2: 5.0},
    n_jobs=-1,
    random_state=SEED
)
rf_model.fit(X_tr_fit, y_tr_fit)
y_pred_rf = rf_model.predict(X_val)
print("=== Random Forest — Val ===")
print(classification_report(y_val, y_pred_rf, target_names=['Low', 'Medium', 'High']))
print(f"✓ Weighted F1 : {f1_score(y_val, y_pred_rf, average='weighted'):.4f}")

# Extra Trees (bonne diversité pour l'ensemble)
et_model = ExtraTreesClassifier(
    n_estimators=500,
    max_depth=20,
    min_samples_leaf=4,
    max_features='sqrt',
    class_weight={0: 1.0, 1: 1.0, 2: 5.0},
    n_jobs=-1,
    random_state=SEED
)
et_model.fit(X_tr_fit, y_tr_fit)
y_pred_et = et_model.predict(X_val)
print("\n=== Extra Trees — Val ===")
print(f"✓ Weighted F1 : {f1_score(y_val, y_pred_et, average='weighted'):.4f}")
```

```python
# ── 6.5 Stacking — Out-Of-Fold méta-features ──────────────────────────────
print("Génération des OOF predictions pour stacking...")

SKF = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)

oof_lgbm = np.zeros((len(X_all), 3))
oof_rf   = np.zeros((len(X_all), 3))
oof_et   = np.zeros((len(X_all), 3))
oof_xgb  = np.zeros((len(X_all), 3)) if XGB_AVAILABLE else None

test_lgbm_folds = np.zeros((len(X_test), 3))
test_rf_folds   = np.zeros((len(X_test), 3))
test_et_folds   = np.zeros((len(X_test), 3))
test_xgb_folds  = np.zeros((len(X_test), 3)) if XGB_AVAILABLE else None

for fold, (tr_idx, oof_idx) in enumerate(SKF.split(X_all, y_all)):
    X_tr, X_oof = X_all.iloc[tr_idx], X_all.iloc[oof_idx]
    y_tr, y_oof = y_all[tr_idx], y_all[oof_idx]
    sw_tr = np.where(y_tr == 2, HIGH_BOOST, 1.0)

    # SMOTE sur ce fold si disponible
    if SMOTE_AVAILABLE:
        try:
            sm = BorderlineSMOTE(
                sampling_strategy={2: int(np.sum(y_tr == 1) * 0.4)},
                k_neighbors=5, random_state=SEED)
            X_tr_s, y_tr_s = sm.fit_resample(X_tr, y_tr)
            sw_tr_s = np.where(y_tr_s == 2, HIGH_BOOST * 0.8, 1.0)
        except:
            X_tr_s, y_tr_s, sw_tr_s = X_tr, y_tr, sw_tr
    else:
        X_tr_s, y_tr_s, sw_tr_s = X_tr, y_tr, sw_tr

    # LightGBM
    m_lgbm = LGBMClassifier(**lgbm_params)
    m_lgbm.fit(X_tr_s, y_tr_s, sample_weight=sw_tr_s,
               eval_set=[(X_oof, y_oof)],
               callbacks=[lgb.early_stopping(150, verbose=False), lgb.log_evaluation(-1)])
    oof_lgbm[oof_idx]  = m_lgbm.predict_proba(X_oof)
    test_lgbm_folds   += m_lgbm.predict_proba(X_test) / SKF.n_splits

    # Random Forest
    m_rf = RandomForestClassifier(n_estimators=300, max_depth=20, min_samples_leaf=4,
                                   max_features='sqrt', class_weight={0:1,1:1,2:5},
                                   n_jobs=-1, random_state=SEED)
    m_rf.fit(X_tr_s, y_tr_s)
    oof_rf[oof_idx]  = m_rf.predict_proba(X_oof)
    test_rf_folds   += m_rf.predict_proba(X_test) / SKF.n_splits

    # Extra Trees
    m_et = ExtraTreesClassifier(n_estimators=300, max_depth=20, min_samples_leaf=4,
                                 max_features='sqrt', class_weight={0:1,1:1,2:5},
                                 n_jobs=-1, random_state=SEED)
    m_et.fit(X_tr_s, y_tr_s)
    oof_et[oof_idx]  = m_et.predict_proba(X_oof)
    test_et_folds   += m_et.predict_proba(X_test) / SKF.n_splits

    # XGBoost
    if XGB_AVAILABLE:
        m_xgb = XGBClassifier(objective='multi:softprob', num_class=3,
                               n_estimators=600, learning_rate=0.05,
                               max_depth=7, subsample=0.85,
                               colsample_bytree=0.75, reg_lambda=1.0,
                               eval_metric='mlogloss', early_stopping_rounds=50,
                               n_jobs=-1, random_state=SEED, verbosity=0)
        m_xgb.fit(X_tr_s, y_tr_s, sample_weight=sw_tr_s,
                  eval_set=[(X_oof, y_oof)], verbose=False)
        oof_xgb[oof_idx]  = m_xgb.predict_proba(X_oof)
        test_xgb_folds   += m_xgb.predict_proba(X_test) / SKF.n_splits

    f1_fold = f1_score(y_oof, np.argmax(oof_lgbm[oof_idx], axis=1), average='weighted')
    print(f"  Fold {fold+1}/5 — LGBM OOF F1: {f1_fold:.4f}")

print(f"\n✓ OOF LGBM Weighted F1 : {f1_score(y_all, np.argmax(oof_lgbm,axis=1), average='weighted'):.4f}")
print(f"✓ OOF RF   Weighted F1 : {f1_score(y_all, np.argmax(oof_rf,  axis=1), average='weighted'):.4f}")
print(f"✓ OOF ET   Weighted F1 : {f1_score(y_all, np.argmax(oof_et,  axis=1), average='weighted'):.4f}")
```

```python
# ── 6.6 Méta-classifieur LightGBM (Level 2) ──────────────────────────────
if XGB_AVAILABLE:
    meta_train = np.hstack([oof_lgbm, oof_rf, oof_et, oof_xgb])
    meta_test  = np.hstack([test_lgbm_folds, test_rf_folds, test_et_folds, test_xgb_folds])
else:
    meta_train = np.hstack([oof_lgbm, oof_rf, oof_et])
    meta_test  = np.hstack([test_lgbm_folds, test_rf_folds, test_et_folds])

# Augmenter les méta-features avec les top features originales
top_feat_names = [c for c in ['global_health_proxy', 'fin_inclusion_score', 'log_turnover',
                               'log_income', 'profit_margin', 'country_te', 'formality_score',
                               'fin_have_now_count', 'total_na_flags', 'efficiency_score',
                               'global_health_sq', 'log_t_over_e']
                  if c in X_all.columns]

meta_train_aug = np.hstack([meta_train, X_all[top_feat_names].values])
meta_test_aug  = np.hstack([meta_test,  X_test[top_feat_names].values])

print(f"Méta-features augmentées train : {meta_train_aug.shape}")

# Méta-classifieur : LightGBM (non-linéaire, plus puissant que LogReg)
meta_lgbm_params = {
    'objective': 'multiclass', 'num_class': 3, 'metric': 'multi_logloss',
    'n_estimators': 500, 'learning_rate': 0.05, 'num_leaves': 31,
    'min_child_samples': 20, 'colsample_bytree': 0.8, 'subsample': 0.8,
    'reg_lambda': 2.0, 'n_jobs': -1, 'random_state': SEED, 'verbose': -1
}

# CV pour le méta-classifieur (anti-overfitting)
skf_meta = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)
oof_meta = np.zeros((len(meta_train_aug), 3))
for fold, (tr_idx, oof_idx) in enumerate(skf_meta.split(meta_train_aug, y_all)):
    m = LGBMClassifier(**meta_lgbm_params)
    m.fit(meta_train_aug[tr_idx], y_all[tr_idx],
          eval_set=[(meta_train_aug[oof_idx], y_all[oof_idx])],
          callbacks=[lgb.early_stopping(50, verbose=False), lgb.log_evaluation(-1)])
    oof_meta[oof_idx] = m.predict_proba(meta_train_aug[oof_idx])

meta_preds_oof = np.argmax(oof_meta, axis=1)
print(f"\n✓ Stack OOF Weighted F1 : {f1_score(y_all, meta_preds_oof, average='weighted'):.4f}")
print(classification_report(y_all, meta_preds_oof, target_names=['Low', 'Medium', 'High']))

# Entraîner méta-clf final
meta_clf = LGBMClassifier(**meta_lgbm_params)
meta_clf.fit(meta_train_aug, y_all, callbacks=[lgb.log_evaluation(-1)])
```

```python
# ── 7.4 Optimisation du seuil de décision pour High ──────────────────────
# L'idée : ajuster le seuil de classification pour High (classe 2)
# afin de maximiser le F1 weighted

lgbm_probas_val = lgbm_model.predict_proba(X_val)  # (N, 3)

best_f1 = 0
best_thresh = 0.5
thresh_results = []

for thresh_high in np.arange(0.10, 0.70, 0.02):
    # Prédire High si proba_high > thresh_high, sinon argmax parmi Low/Medium
    preds_adj = []
    for proba in lgbm_probas_val:
        if proba[2] >= thresh_high:
            preds_adj.append(2)
        else:
            preds_adj.append(np.argmax(proba[:2]))
    f1 = f1_score(y_val, preds_adj, average='weighted')
    thresh_results.append((thresh_high, f1))
    if f1 > best_f1:
        best_f1 = f1
        best_thresh = thresh_high

thresh_df = pd.DataFrame(thresh_results, columns=['threshold_high', 'weighted_f1'])

fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(thresh_df['threshold_high'], thresh_df['weighted_f1'], 'b-o', ms=4)
ax.axvline(best_thresh, color='red', ls='--', label=f'Optimal = {best_thresh:.2f}')
ax.axhline(best_f1, color='green', ls='--', label=f'Best F1 = {best_f1:.4f}')
ax.set_xlabel('Seuil de décision pour classe High')
ax.set_ylabel('Weighted F1')
ax.set_title('Optimisation du seuil de décision pour High', fontweight='bold')
ax.legend()
plt.tight_layout()
plt.show()

print(f"\nMeilleur seuil High : {best_thresh:.2f} → F1 = {best_f1:.4f}")
```

# 8. Entraînement Final & Soumission

```python
# ── 8.1 Sélectionner la meilleure stratégie ────────────────────────────────
# Comparer : Soft Voting vs Stack vs LGBM+threshold
scores_compare = {}

# Soft Voting
scores_compare['Soft Voting'] = f1_score(y_val, soft_vote_val, average='weighted')

# LGBM + threshold optimisé
preds_thresh = []
for proba in lgbm_probas_val:
    if proba[2] >= best_thresh:
        preds_thresh.append(2)
    else:
        preds_thresh.append(np.argmax(proba[:2]))
scores_compare['LGBM + Threshold'] = f1_score(y_val, preds_thresh, average='weighted')

# Blending pondéré : donner plus de poids au LGBM (généralement le meilleur)
if XGB_AVAILABLE and CB_AVAILABLE:
    w_lgbm, w_xgb, w_rf, w_cb = 0.40, 0.25, 0.10, 0.25
    blend_val = (w_lgbm * lgbm_model.predict_proba(X_val)
                + w_xgb * xgb_model.predict_proba(X_val)
                + w_rf  * rf_model.predict_proba(X_val)
                + w_cb  * cb_model.predict_proba(X_val))
elif XGB_AVAILABLE:
    w_lgbm, w_xgb, w_rf = 0.50, 0.30, 0.20
    blend_val = (w_lgbm * lgbm_model.predict_proba(X_val)
                + w_xgb * xgb_model.predict_proba(X_val)
                + w_rf  * rf_model.predict_proba(X_val))
else:
    blend_val = lgbm_model.predict_proba(X_val)

blend_preds_val = np.argmax(blend_val, axis=1)
scores_compare['Weighted Blending'] = f1_score(y_val, blend_preds_val, average='weighted')

print("\n=== Comparaison finale ===")
for name, score in sorted(scores_compare.items(), key=lambda x: -x[1]):
    marker = '✓ BEST' if score == max(scores_compare.values()) else ''
    print(f"  {name:25s}: {score:.4f}  {marker}")

best_strategy = max(scores_compare, key=scores_compare.get)
print(f"\n→ Stratégie retenue : {best_strategy}")
```

```python
# ── 8.2 Entraînement final sur TOUT le train (+ pseudo-labels) ────────────
print("Entraînement final...")

X_final = X_all_aug if PSEUDO_AVAILABLE else X_all
y_final = y_all_aug if PSEUDO_AVAILABLE else y_all
sw_final = np.where(y_final == 2, HIGH_BOOST, 1.0)

# SMOTE sur le dataset final
if SMOTE_AVAILABLE:
    try:
        sm_final = BorderlineSMOTE(
            sampling_strategy={2: int(np.sum(y_final == 1) * 0.4)},
            k_neighbors=5, random_state=SEED)
        X_final_sm, y_final_sm = sm_final.fit_resample(X_final, y_final)
        sw_final_sm = np.where(y_final_sm == 2, HIGH_BOOST * 0.8, 1.0)
        print(f"  SMOTE final : {len(X_final)} → {len(X_final_sm)}")
    except:
        X_final_sm, y_final_sm, sw_final_sm = X_final, y_final, sw_final
else:
    X_final_sm, y_final_sm, sw_final_sm = X_final, y_final, sw_final

# LightGBM final
n_est_final = (lgbm_model.best_iteration_ + 100) if hasattr(lgbm_model, 'best_iteration_') else 1500
lgbm_params_final = dict(lgbm_params)
lgbm_params_final['n_estimators'] = n_est_final

lgbm_final = LGBMClassifier(**lgbm_params_final)
lgbm_final.fit(X_final_sm, y_final_sm, sample_weight=sw_final_sm,
               callbacks=[lgb.log_evaluation(-1)])

# RF final
rf_final = RandomForestClassifier(
    n_estimators=600, max_depth=20, min_samples_leaf=4,
    max_features='sqrt', class_weight={0:1,1:1,2:5}, n_jobs=-1, random_state=SEED)
rf_final.fit(X_final_sm, y_final_sm)

# ET final
et_final = ExtraTreesClassifier(
    n_estimators=500, max_depth=20, min_samples_leaf=4,
    max_features='sqrt', class_weight={0:1,1:1,2:5}, n_jobs=-1, random_state=SEED)
et_final.fit(X_final_sm, y_final_sm)

# XGBoost final
if XGB_AVAILABLE:
    n_xgb_final = xgb_model.best_iteration + 100 if hasattr(xgb_model, 'best_iteration') else 600
    xgb_final = XGBClassifier(
        objective='multi:softprob', num_class=3,
        n_estimators=n_xgb_final, learning_rate=0.03,
        max_depth=7, subsample=0.85, colsample_bytree=0.75,
        reg_lambda=1.0, n_jobs=-1, random_state=SEED, verbosity=0)
    xgb_final.fit(X_final_sm, y_final_sm, sample_weight=sw_final_sm, verbose=False)

# CatBoost final
if CB_AVAILABLE:
    n_cb_final = cb_model.best_iteration_ + 50 if hasattr(cb_model, 'best_iteration_') else 600
    cb_final = CatBoostClassifier(
        iterations=n_cb_final, learning_rate=0.03, depth=8, l2_leaf_reg=3,
        class_weights=[w_low, w_med, w_high], random_seed=SEED, verbose=False)
    cb_final.fit(X_final_sm, y_final_sm)

print("✓ Entraînement final terminé")
```

```python
# ── 8.3 Prédictions finales : Stack + Blend + Threshold ──────────────────
print(f"Génération des prédictions finales...")

# Probas base models
test_proba_lgbm = lgbm_final.predict_proba(X_test)
test_proba_rf   = rf_final.predict_proba(X_test)
test_proba_et   = et_final.predict_proba(X_test)

if XGB_AVAILABLE:
    test_proba_xgb  = xgb_final.predict_proba(X_test)
    meta_test_final = np.hstack([test_proba_lgbm, test_proba_rf, test_proba_et, test_proba_xgb])
else:
    meta_test_final = np.hstack([test_proba_lgbm, test_proba_rf, test_proba_et])

meta_test_final_aug = np.hstack([meta_test_final, X_test[top_feat_names].values])

# Prédictions stack méta-classifieur
stack_proba_test = meta_clf.predict_proba(meta_test_final_aug)

# Blending direct
if XGB_AVAILABLE and CB_AVAILABLE:
    w_lgbm, w_xgb, w_rf, w_cb, w_et = 0.35, 0.20, 0.10, 0.20, 0.15
    direct_blend = (w_lgbm * test_proba_lgbm + w_xgb * test_proba_xgb
                  + w_rf  * test_proba_rf   + w_cb  * cb_final.predict_proba(X_test)
                  + w_et  * test_proba_et)
elif XGB_AVAILABLE:
    w_lgbm, w_xgb, w_rf, w_et = 0.40, 0.25, 0.15, 0.20
    direct_blend = (w_lgbm * test_proba_lgbm + w_xgb * test_proba_xgb
                  + w_rf  * test_proba_rf   + w_et  * test_proba_et)
else:
    direct_blend = 0.6 * test_proba_lgbm + 0.4 * test_proba_et

# Blend final : 60% stack + 40% direct
final_proba = 0.6 * stack_proba_test + 0.4 * direct_blend

# Appliquer le seuil optimisé pour High
final_preds_enc = []
for proba in final_proba:
    if proba[2] >= best_thresh:
        final_preds_enc.append(2)
    else:
        final_preds_enc.append(np.argmax(proba[:2]))

final_preds = [inv_label_map[p] for p in final_preds_enc]

print("\nDistribution des prédictions test:")
uniq, cnts = np.unique(final_preds, return_counts=True)
for u, c in zip(uniq, cnts):
    print(f"  {u:8s}: {c:4d} ({100*c/len(final_preds):.1f}%)")

submission = pd.DataFrame({'ID': test_raw['ID'].values, 'Target': final_preds})
submission.to_csv('submission.csv', index=False)
print(f"\n✓ submission.csv sauvegardé ({len(submission)} lignes)")
submission.head()
```

```python
# ── 8.0 Pseudo-labeling (exemples test haute confiance) ─────────────────
# Stratégie : ajouter au train les prédictions test avec confiance >= 90%
# → le modèle final "apprend" aussi depuis le test set non labelisé

PSEUDO_THRESHOLD = 0.90

# Prédictions test avec le modèle stack
test_proba_stack = meta_clf.predict_proba(meta_test_aug)
test_max_proba   = test_proba_stack.max(axis=1)
test_pseudo_mask = test_max_proba >= PSEUDO_THRESHOLD

print(f"Exemples test haute confiance (>= {PSEUDO_THRESHOLD}) : {test_pseudo_mask.sum()} / {len(X_test)}")
pseudo_labels = np.argmax(test_proba_stack[test_pseudo_mask], axis=1)
print("Distribution des pseudo-labels :")
for cls, name in inv_label_map.items():
    n = (pseudo_labels == cls).sum()
    print(f"  {name:8s}: {n:4d} ({100*n/max(1, len(pseudo_labels)):.1f}%)")

if test_pseudo_mask.sum() > 50:
    X_pseudo    = X_test[test_pseudo_mask]
    y_pseudo    = pseudo_labels
    X_all_aug   = pd.concat([X_all, X_pseudo], ignore_index=True)
    y_all_aug   = np.concatenate([y_all, y_pseudo])
    print(f"\nTrain augmenté : {len(X_all)} → {len(X_all_aug)} exemples")
    PSEUDO_AVAILABLE = True
else:
    X_all_aug, y_all_aug = X_all.copy(), y_all.copy()
    PSEUDO_AVAILABLE = False
    print("⚠ Pas assez d'exemples — pseudo-labeling ignoré")
```

| Problème identifié | Solution v3 | Impact attendu |
|---|---|---|

| Ensemble naïf (poids égaux) | **Weighted Blending (LGBM=40%, XGB=25%...)** | +0.005 |
| Feature engineering insuffisant | **Scores composites avancés + interactions pays** | +0.01 |
