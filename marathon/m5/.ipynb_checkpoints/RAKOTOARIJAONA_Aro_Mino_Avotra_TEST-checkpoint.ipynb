# Problem framing

> ## Main goal
 > The goal of this challenge is to predict the probability that a farmer will adopt a practice after their first training session delivered by DigiCow. The predictions must be based on data that would be available at the time of the training event. Accurately predicting adoption can help DigiCow:
> * Prioritise follow‑up support for farmers who are most likely to adopt new practices.
> * Tailor extension services and advice based on predicted adoption likelihood.
> * Improve training programmes to increase real‑world impact over time.

> ## Dataset and prediction task
> You are given a dataset where each row represents a farmer’s first training event and a set of features that describe the farmer and their context at that time. The task is to build a model that outputs the following:
> * Probability of adoption within 7 days
> * Probability of adoption within 90 days
> * Probability of adoption within 120 days
>  
> The model must learn from information that a farmer would realistically have at the time of their training—features that occur after these time horizons or leak future information should be avoided.

> ## Can we solve it without ML?
> Traditional rule‑based systems (e.g. simple thresholds on farmer age or location) are limited and cannot reliably model the complex patterns underlying farmer behaviour. In contrast, machine learning models allow us to:
> 
> * Capture complex, non‑linear relationships: Farmer adoption decisions are influenced by many interacting factors (e.g., experience, training type, prior behaviours). ML captures these patterns automatically from data.
> * redict probabilities rather than hard classes: The evaluation favours probability outputs (not just yes/no predictions), and many ML models excel at producing well‑calibrated probability estimates.
> * Generalise to new examples: Unlike static rules, machine learning adapts to unseen data, making it more reliable when applied to new sets of farmers or changing contexts.
> * Handle imbalanced outcomes: Since adoption might be rarer than non‑adoption, ML models can be tuned to deal with class imbalance and still learn meaningful patterns.


> ## Technical formulation
> 
> * Problem type: Supervised Machine Learning
> * Task type: Task type: Multi‑label classification with probabilistic outputs
> * Input features (X): Processed features representing farmer demographics, training details, and other relevant variables known at training time
> * Output (y): Three probability scores (for adoption within 7, 90 and 120 days respectively)
> * 
> The challenge uses a multi‑metric evaluation, where your model’s performance is assessed using:
> * Log Loss (75%) – Encourages models to give reliable probability estimates. Predictions that are confidently wrong are penalised heavily.
> * ROC‑AUC (25%) – Measures how well the model ranks true adopters above non‑adopters across thresholds, which is especially valuable when the data is imbalanced.
>
> The final score is a weighted average of these metrics across all three adoption horizons (7, 90 and 120 days).

```python
print("\n" + "=" * 60)
print("MISSING VALUES ANALYSIS")
print("=" * 60)

def analyze_missing(df, name):
    missing = df.isnull().sum()
    missing_pct = (missing / len(df)) * 100
    missing_df = pd.DataFrame({
        'Column': missing.index,
        'Missing_Count': missing.values,
        'Percentage': missing_pct.values
    }).query('Missing_Count > 0').sort_values('Percentage', ascending=False)
    
    print(f"\n{name} - Columns with Missing Values:")
    if len(missing_df) > 0:
        print(missing_df.to_string(index=False))
        
        # Visualize top missing columns
        if len(missing_df) > 0:
            top_missing = missing_df.head(15)
            plt.figure(figsize=(12, 6))
            plt.barh(top_missing['Column'], top_missing['Percentage'], color='coral')
            plt.xlabel('Missing Percentage (%)')
            plt.title(f'{name} - Top Columns with Missing Values', fontsize=14, fontweight='bold')
            plt.gca().invert_yaxis()
            plt.grid(axis='x', alpha=0.3)
            plt.tight_layout()
            plt.show()
    else:
        print("No missing values found!")
    
    return missing_df

train_missing = analyze_missing(train_df, "TRAIN SET")
test_missing = analyze_missing(test_df, "TEST SET")
```

> ## Data transformation pipeline

```python
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler, LabelEncoder
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
import pandas as pd
import numpy as np

def change_object_to_cat(df):
  # changes objects columns to category

  df = df.copy()
  list_str_obj_cols = df.columns[df.dtypes == "object"].tolist()
  for str_obj_col in list_str_obj_cols:
      df[str_obj_col] = df[str_obj_col].astype("category")

  return df,list_str_obj_cols

class FeatureEngineering(BaseEstimator, TransformerMixin):
    """
    Crée des features dérivées pour booster le modèle
    """
    def __init__(self):
        pass
    
    def fit(self, X, y=None):
        return self
    
    def transform(self, X):
        X = X.copy()
        
        # Profit Margin
        X['profit_margin'] = np.where(
            (X['personal_income'].notna()) & (X['business_expenses'].notna()) & (X['personal_income'] != 0),
            (X['personal_income'] - X['business_expenses']) / X['personal_income'],
            np.nan
        )
        X['profit_margin'] = X['profit_margin'].clip(-1, 1)
        
        # Financial Access Score
        financial_features = ['has_bank_account', 'has_loan_account', 'has_internet_banking',
                              'has_debit_card', 'medical_insurance', 'funeral_insurance']
        available_features = [f for f in financial_features if f in X.columns]
        
        def calc_fin_score(row):
            score = 0
            valid = 0
            for f in available_features:
                val = row[f]
                if pd.notna(val):
                    valid += 1
                    if str(val).lower() in ['yes', 'have now', 'have']:
                        score += 1
                    elif str(val).lower() in ["used to have", "used to have but don't have now"]:
                        score += 0.5
            return score / valid if valid > 0 else np.nan
        
        X['financial_access_score'] = X.apply(calc_fin_score, axis=1)
        
        # Age Bins
        bins = [0, 25, 35, 45, 55, 65, 100]
        labels = ['<25','25-34','35-44','45-54','55-64','65+']
        X['owner_age_bin'] = pd.cut(X['owner_age'], bins=bins, labels=labels)
        
        # Income-to-Expenses Ratio
        X['income_expenses_ratio'] = np.where(
            (X['personal_income'].notna()) & (X['business_expenses'].notna()) & (X['business_expenses'] != 0),
            X['personal_income'] / X['business_expenses'],
            np.nan
        )
        
        # Business Maturity
        X['business_maturity'] = pd.cut(
            X['business_age_years'],
            bins=[-1,1,3,5,100],
            labels=['new','young','established','mature']
        )
        
        # Risk Perception Score
        attitude_cols = ['attitude_worried_shutdown', 'attitude_more_successful_next_year',
                         'attitude_stable_business_environment', 'attitude_satisfied_with_achievement']
        attitude_cols = [c for c in attitude_cols if c in X.columns]
        
        def calc_risk(row):
            score = 0
            count = 0
            for c in attitude_cols:
                val = str(row[c]).lower() if pd.notna(row[c]) else 'unknown'
                if val in ['yes','have now','have']:
                    score += 1
                elif val in ['no','never']:
                    score += 0
                count += 1
            return score / count if count > 0 else np.nan
        
        X['risk_perception_score'] = X.apply(calc_risk, axis=1)
        
        # Country-Specific Score
        if 'country' in X.columns:
            country_map = {c:i for i,c in enumerate(sorted(X['country'].dropna().unique()))}
            X['country_score'] = X['country'].map(country_map)
        
        # Insurance Coverage Index
        insurance_cols = ['has_insurance', 'medical_insurance', 'funeral_insurance']
        insurance_cols = [c for c in insurance_cols if c in X.columns]
        X['insurance_coverage_index'] = X[insurance_cols].apply(
            lambda row: sum([1 if str(v).lower() in ['yes','have now'] else 0 for v in row])/len(row) 
            if row.notna().any() else np.nan, axis=1
        )
        
        # Digital Banking Adoption Flag
        digital_cols = ['has_internet_banking','has_debit_card','has_loan_account','has_bank_account']
        digital_cols = [c for c in digital_cols if c in X.columns]
        X['digital_banking_flag'] = X[digital_cols].apply(
            lambda row: 1 if any([str(v).lower() in ['yes','have now'] for v in row]) else 0, axis=1
        )
        
        return X

class NumericMissingFlag(BaseEstimator, TransformerMixin):
    """
    Ajoute un flag '_is_missing' pour les colonnes numériques avec N/A
    """
    def __init__(self, columns=None, threshold=0.05):
        self.columns = columns
        self.threshold = threshold
        self.flags_cols_ = []

    def fit(self, X, y=None):
        X = X.copy()
        self.flags_cols_ = []
        for col in (self.columns if self.columns is not None else X.columns):
            if X[col].isna().mean() >= self.threshold:
                self.flags_cols_.append(col)
        return self
        
    def set_output(self, transform=None):
        return self

    def transform(self, X):
        X = X.copy()
        for col in self.flags_cols_:
            X[col + "_is_missing"] = X[col].isna().astype(int)
        return X

class CategoricalCleaner(BaseEstimator, TransformerMixin):
    def __init__(self, high_missing_threshold=0.4, low_missing_threshold=0.05):
        self.high_missing_threshold = high_missing_threshold
        self.low_missing_threshold = low_missing_threshold
        self.flags_cols_ = []
        self.modes_ = {}
        
    def fit(self, X, y=None):
        X = X.copy()
        for col in X.columns:
            X[col] = X[col].apply(self.clean_value)
            missing_ratio = X[col].isna().mean()
            # Flag si beaucoup de NaN
            if missing_ratio > self.high_missing_threshold:
                self.flags_cols_.append(col + "_is_missing")
            # Mode pour faible missing
            if missing_ratio <= self.low_missing_threshold:
                self.modes_[col] = X[col].mode()[0]
            # Remplissage sécurité pour ratio intermédiaire
            elif missing_ratio > self.low_missing_threshold:
                X[col] = X[col].fillna("unknown")
        return self
    
    def transform(self, X):
        X = X.copy()
        for col in X.columns:
            X[col] = X[col].apply(self.clean_value)
            missing_ratio = X[col].isna().mean()
            # flag et fill pour high_missing
            if missing_ratio > self.high_missing_threshold:
                X[col + "_is_missing"] = X[col].isna().astype(int)
                X[col] = X[col].fillna("unknown")
            # fill pour ratio intermédiaire
            elif missing_ratio > self.low_missing_threshold:
                X[col] = X[col].fillna("unknown")
            # fill avec mode pour faible missing
            else:
                X[col] = X[col].fillna(self.modes_[col])
        
        return X
        
    def set_output(self, transform=None):
        return self
    
    @staticmethod
    def clean_value(val):
        if pd.isna(val):
            return "unknown"
        val = str(val).strip()
        val = val.replace("’", "'")
        val = " ".join(val.split())
        val = val.lower()
        if val in ["don’t know", "don't know", "don’t know or n/a", "don't know (do not show)",
                   "don’t know / doesn’t apply", "refused", "n/a", "do not know / n‎/a", "na"]:
            return "unknown"
        return val


categorical_cols = [
    'country', 'attitude_stable_business_environment', 'attitude_worried_shutdown', 
    'compliance_income_tax', 'perception_insurance_doesnt_cover_losses', 
    'perception_cannot_afford_insurance', 'motor_vehicle_insurance', 'has_mobile_money', 
    'current_problem_cash_flow', 'has_cellphone', 'owner_sex', 'offers_credit_to_customers', 
    'attitude_satisfied_with_achievement', 'has_credit_card', 'keeps_financial_records', 
    'perception_insurance_companies_dont_insure_businesses_like_yours', 
    'perception_insurance_important', 'has_insurance', 'covid_essential_service', 
    'attitude_more_successful_next_year', 'problem_sourcing_money', 'marketing_word_of_mouth', 
    'has_loan_account', 'has_internet_banking', 'has_debit_card', 'future_risk_theft_stock', 
    'medical_insurance', 'funeral_insurance', 'motivation_make_more_money', 
    'uses_friends_family_savings', 'uses_informal_lender'
]

numerical_cols = ['owner_age', 'personal_income', 'business_expenses', 'business_turnover', 
                  'business_age_years', 'business_age_months']

new_numeric_features = [
    'profit_margin', 'financial_access_score', 'income_expenses_ratio',
    'risk_perception_score', 'insurance_coverage_index', 'country_score'
]

new_categorical_features = [
    'owner_age_bin', 'business_maturity', 'digital_banking_flag'
]

numeric_pipeline_cols = numerical_cols + new_numeric_features

categorical_pipeline_cols = categorical_cols + new_categorical_features

# Pipelines
categorical_pipeline = Pipeline([
    ('cleaner', CategoricalCleaner()),
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

numeric_pipeline = Pipeline([
    ('missing_flag', NumericMissingFlag(columns=numerical_cols, threshold=0.05)),
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

preprocessor = ColumnTransformer([
    ('num', numeric_pipeline, numeric_pipeline_cols),
    ('cat', categorical_pipeline, categorical_pipeline_cols)
])

preprocessor.set_output(transform="pandas")

full_pipeline = Pipeline([
    ('feature_engineering', FeatureEngineering()),
    ('preprocessor', preprocessor)
])
```

```python
le = LabelEncoder()
y_train = le.fit_transform(y_t)
y_val   = le.transform(y_v)

X_train = full_pipeline.fit_transform(X_t)
X_val  = full_pipeline.transform(X_v)
X_test  = full_pipeline.transform(test_df)

print("Shape train après pipeline:", X_train.shape)
print("Shape val après pipeline :", X_val.shape)
print("Shape test après pipeline :", X_test.shape)
```

```python
# Model selectionned for ensemble learning

lgbm = LGBMClassifier(
    boosting_type='gbdt',
    class_weight=None,
    learning_rate=0.1,
    max_depth=20,
    min_child_samples=50,
    n_estimators=100,
    num_leaves=63,
    objective='multiclass',
    random_state=42,
    verbose=-1
)

xgb = XGBClassifier(
    objective='multi:softprob',
    learning_rate=0.1,
    max_depth=3,
    n_estimators=100,
    subsample=0.8,
    colsample_bytree=1,
    eval_metric='logloss',
    use_label_encoder=False,
    random_state=42
)

rf_normal = RandomForestClassifier(
    n_estimators=200,
    class_weight=None,
    random_state=42
)
```

```python
from sklearn.ensemble import VotingClassifier

voting_clf = VotingClassifier(
    estimators=[
        ('lgbm', lgbm),
        ('xgb', xgb),
        ('rf_norm', rf_normal),
    ],
    voting='soft',
    n_jobs=-1
)

voting_clf.fit(X_train, y_train)
y_pred = voting_clf.predict(X_train)
print("Train : \n", classification_report(y_train, y_pred))
y_pred = voting_clf.predict(X_val)
print("Val : \n", classification_report(y_val, y_pred))
```

```python
from sklearn.ensemble import StackingClassifier
from sklearn.linear_model import LogisticRegression

stacking_clf = StackingClassifier(
    estimators=[
        ('lgbm', lgbm),
        ('xgb', xgb),
        ('rf_norm', rf_normal),

    ],
    final_estimator=LogisticRegression(),
    stack_method='predict_proba',
    cv=5,
    n_jobs=-1
)

stacking_clf.fit(X_train, y_train)
y_pred = stacking_clf.predict(X_train)
print("Train : \n", classification_report(y_train, y_pred))
y_pred = stacking_clf.predict(X_val)
print("Val : \n", classification_report(y_val, y_pred))
```
